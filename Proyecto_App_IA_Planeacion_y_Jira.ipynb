## 1. Caso de negocio

> Recordatorio de la Sesión 6: el **caso de negocio justifica** la inversión (¿vale la pena hacerlo?), el
> **acta constitutiva autoriza** el proyecto.

### 1.1 Necesidad del negocio

El área de Servicio al Cliente recibe alrededor de **6.000 consultas mensuales** por canal web y correo.
El análisis preliminar muestra que:

- El **68 %** de las consultas son repetitivas (estado de pedido, políticas de devolución, horarios, precios).
- El **tiempo promedio de primera respuesta** es de **6 horas** y la meta del área es de 1 hora.
- El equipo de 8 agentes está saturado y la rotación anual supera el 30 %.
- El **12 %** de los clientes que esperan más de 4 horas abandona la compra.

### 1.2 Oportunidad

Un asistente conversacional con IA puede responder automáticamente las consultas repetitivas, con acceso a
la base documental de la empresa, y derivar al agente humano solo los casos complejos.

### 1.3 Alternativas evaluadas

| Alternativa | Costo estimado | Ventajas | Desventajas | Decisión |
|---|---|---|---|---|
| **A. No hacer nada** | $0 | Sin inversión | La saturación crece con las ventas; se pierden clientes | Descartada |
| **B. Contratar 4 agentes más** | ~$240 M/año | Rápido de implementar | Costo recurrente alto; no escala; no resuelve el horario nocturno | Descartada |
| **C. Chatbot de reglas** | ~$60 M | Económico | Rígido; no entiende lenguaje natural; alta tasa de abandono | Descartada |
| **D. Asistente con IA generativa + base documental** | ~$180 M | Escala, atiende 24/7, aprende del corpus de la empresa | Requiere gobierno de datos y control de calidad de respuestas | **Seleccionada** |

### 1.4 Beneficios esperados

| Beneficio | Indicador | Línea base | Meta a 6 meses |
|---|---|---|---|
| Respuesta más rápida | Tiempo de primera respuesta | 6 horas | 2 minutos en consultas automatizadas |
| Menos carga operativa | Consultas resueltas sin agente | 0 % | 55 % |
| Disponibilidad | Cobertura horaria | 8×5 | 24×7 |
| Satisfacción | CSAT | 3,4 / 5 | 4,2 / 5 |
| Ahorro operativo | Costo por consulta atendida | $4.200 | $1.900 |

### 1.5 Viabilidad

| Dimensión | Evaluación |
|---|---|
| Técnica | Viable: existen modelos y servicios de IA maduros; el equipo tiene experiencia en desarrollo web |
| Económica | Viable: ver el cálculo de VPN, ROI y periodo de recuperación en la sección 9 |
| Operativa | Requiere gestión del cambio con los agentes; se incluye la épica de adopción |
| Legal | Debe cumplir la Ley 1581 de 2012 de protección de datos personales; se incluye la épica de IA responsable |
| Cronograma | Viable en 12 semanas con el equipo definido |

## 2. Acta constitutiva (*Project Charter*)

### 2.1 Identificación

| Campo | Valor |
|---|---|
| **Nombre del proyecto** | AsistIA — Asistente de IA para Servicio al Cliente |
| **Código en Jira** | `ASIST` (clave del espacio) |
| **Patrocinador** | Gerente de Operaciones |
| **Gerente del proyecto** | Estudiante a cargo |
| **Fecha de inicio** | Semana 1 |
| **Duración** | 12 semanas (6 sprints de 2 semanas) |
| **Presupuesto autorizado** | $180.000.000 COP |

### 2.2 Propósito

Desarrollar e implantar un asistente conversacional con inteligencia artificial que atienda de forma
automática las consultas repetitivas de los clientes, integrado al canal web y al sistema de tickets,
con derivación a agente humano cuando el caso lo requiera.

### 2.3 Objetivos medibles y criterios de éxito

| # | Objetivo (SMART) | Métrica | Meta | Plazo |
|---|---|---|---|---|
| 1 | Automatizar la atención de consultas repetitivas | % de consultas resueltas sin agente | ≥ 55 % | Semana 12 |
| 2 | Reducir el tiempo de primera respuesta | Minutos promedio | ≤ 2 min en consultas automatizadas | Semana 12 |
| 3 | Mantener la calidad de las respuestas | % de respuestas correctas en el conjunto de evaluación | ≥ 85 % | Semana 10 |
| 4 | No degradar la satisfacción | CSAT del canal | ≥ 4,0 / 5 | Semana 12 |
| 5 | Cumplir el presupuesto | Desviación de costo | ≤ 10 % | Semana 12 |

### 2.4 Alcance

| Incluido | **Excluido** |
|---|---|
| Asistente conversacional sobre la base documental de la empresa | Integración con WhatsApp y redes sociales |
| Integración con el canal web y el sistema de tickets | Atención en idiomas distintos al español |
| Derivación a agente humano | Automatización de procesos de facturación o pagos |
| Panel de métricas de uso y calidad | Reemplazo del sistema de tickets actual |
| Capacitación a los agentes | Desarrollo de una aplicación móvil nativa |

### 2.5 Hitos principales

| Hito | Semana | Criterio de cumplimiento |
|---|---|---|
| H1 — Planeación aprobada y entorno listo | 2 | Acta firmada, tablero configurado, ambientes creados |
| H2 — Corpus documental cargado y depurado | 4 | Base de conocimiento con documentos procesados y validados |
| H3 — Prototipo funcional evaluado | 6 | Prototipo responde el conjunto de evaluación con ≥ 70 % de acierto |
| H4 — Integración con canal web y tickets | 10 | Asistente operando en ambiente de pruebas de punta a punta |
| H5 — Piloto y paso a producción | 12 | Piloto con usuarios reales superado y despliegue autorizado |

### 2.6 Riesgos de alto nivel

1. Calidad insuficiente de la base documental.
2. Respuestas incorrectas o inventadas por el modelo frente al cliente.
3. Incumplimiento normativo en el tratamiento de datos personales.
4. Resistencia del equipo de agentes al cambio.

### 2.7 Criterios de salida

El proyecto se cierra cuando: el asistente opera en producción, se cumplen los objetivos 1 a 4, la
documentación de operación está entregada, los agentes están capacitados y el patrocinador firma la
aceptación final.

### 2.8 Autoridad del gerente del proyecto

El gerente del proyecto queda autorizado para asignar los recursos del equipo, decidir sobre el contenido
de cada sprint y aprobar cambios dentro de los límites definidos en el plan de gobernanza (sección 4).

### 3.1 Registro de interesados

### 3.2 Matriz RACI

**R** = Encargado (ejecuta) · **A** = Responsable (rinde cuentas) · **C** = Consultado · **I** = Informado

| Actividad | Patroc. | PM | PO | SM | Equipo Dev | TI | Legal |
|---|:---:|:---:|:---:|:---:|:---:|:---:|:---:|
| Aprobación del acta constitutiva | **A** | R | C | I | I | C | I |
| Priorización del backlog | I | C | **A/R** | I | C | I | I |
| Definición de la arquitectura | I | C | I | I | **A/R** | C | I |
| Curaduría del corpus documental | I | C | **A** | I | R | I | C |
| Desarrollo del asistente | I | I | C | I | **A/R** | I | I |
| Eliminación de impedimentos | I | C | I | **A/R** | C | C | I |
| Revisión de privacidad y datos personales | I | C | C | I | C | I | **A/R** |
| Pruebas y aceptación funcional | I | C | **A** | I | R | I | I |
| Despliegue a producción | **A** | R | C | I | R | C | I |
| Capacitación a los agentes | I | **A/R** | C | I | C | I | I |
| Cierre del proyecto | **A** | R | C | C | I | I | I |

Cada fila tiene **una sola A**, tal como exige la regla vista en la sesión de gobernanza. La validación
automática de esta matriz está en la sección 10.

## 4. Plan de gobernanza

Estructura vista en la sesión de Gobernanza: **Políticas → Límites → Escalamiento**.

### 4.1 Políticas

| Política | Regla |
|---|---|
| Definición de terminado (DoD) | Código revisado por un par, pruebas pasando, documentación actualizada y aceptado por el PO |
| Control de cambios | Toda solicitud entra como incidencia tipo *Solicitud de cambio* en Jira y es evaluada en la revisión de sprint |
| Calidad de respuestas | Ninguna versión pasa a producción con menos de 85 % de acierto en el conjunto de evaluación |
| Datos personales | Prohibido usar datos de clientes sin anonimizar en pruebas o entrenamiento |
| Transparencia | El usuario siempre sabe que habla con un asistente de IA y puede pedir un agente humano |
| Reportes | Informe de avance al patrocinador al cierre de cada sprint |

### 4.2 Límites de decisión

| Variable | Decide el equipo / PM | Escala al PO y al comité | Escala al patrocinador |
|---|---|---|---|
| Desviación de costo | Hasta 5 % | Entre 5 % y 10 % | Más del 10 % |
| Desviación de cronograma | Hasta 3 días | De 3 días a 2 semanas | Más de 2 semanas o afecta un hito |
| Cambio de alcance | Reordenar el backlog | Nueva funcionalidad dentro del objetivo | Cambio del objetivo del proyecto |
| Riesgo | Severidad baja | Severidad media | Severidad alta o crítica |
| Incidente de datos personales | — | — | Inmediato, junto con el oficial de protección de datos |

### 4.3 Escalamiento

| Disparador | Destino | Tiempo máximo de respuesta | Información mínima |
|---|---|---|---|
| Impedimento sin resolver por más de 48 horas | Scrum Master → PM | 1 día | Impedimento, impacto y alternativas |
| Desviación de costo mayor al 10 % | Patrocinador | 3 días | Causa, impacto financiero y recomendación |
| Riesgo de severidad alta materializado | Patrocinador y comité | 1 día | Riesgo, impacto, plan de contingencia |
| Hallazgo legal sobre datos personales | Oficial de datos y patrocinador | Inmediato | Hallazgo, alcance y medidas tomadas |

### 4.4 Ritmo de gobierno

| Ceremonia | Frecuencia | Duración | Participantes | Propósito de gobernanza |
|---|---|---|---|---|
| Reunión diaria | Diaria | 15 min | Equipo, SM | Autoorganización y detección de bloqueos |
| Planificación de sprint | Cada 2 semanas | 2 h | Equipo, PO, SM | Compromiso dentro de los límites |
| Revisión de sprint | Cada 2 semanas | 1 h | Equipo, PO, interesados | Transparencia y aceptación del incremento |
| Retrospectiva | Cada 2 semanas | 1 h | Equipo, SM | Mejora del proceso |
| Comité de seguimiento | Mensual | 1 h | Patrocinador, PM, PO | Decisiones que superan los límites del equipo |

estrictamente técnicas: un proyecto de IA es mucho más que entrenar un modelo.

```

   │
   ├── E1. Descubrimiento y caso de negocio
   ├── E2. Datos: fuentes, calidad y gobierno
   ├── E3. Modelo de IA: línea base y evaluación
```

| 2 | 3–4 | Corpus documental y selección del enfoque técnico | H2 |
| 3 | 5–6 | Prototipo, evaluación y experiencia de usuario | H3 |

````python
# Chequeo de capacidad por sprint

VELOCIDAD = 32  # puntos por sprint estimados para el equipo
SPRINTS = range(1, 7)

print(f"{'Sprint':<8}{'Historias':>10}{'Puntos':>8}{'Capacidad':>11}   Estado")
print("-" * 78)
sobrecargados = []
for s in SPRINTS:
    hist = [h for h in BACKLOG if h[5] == s]
    pts = sum(h[4] for h in hist)
    if pts > VELOCIDAD:
        estado = f"SOBRECARGADO (+{pts - VELOCIDAD} pts)"
        sobrecargados.append(s)
    elif pts < VELOCIDAD * 0.8:
        estado = "holgura disponible"
    else:
        estado = "OK"
    barra = "#" * pts
    print(f"{s:<8}{len(hist):>10}{pts:>8}{VELOCIDAD:>11}   {estado}")
    print(f"{'':<8}{barra}")
print("-" * 78)
print(f"Total: {sum(h[4] for h in BACKLOG)} puntos en {len(list(SPRINTS))} sprints "
      f"(capacidad total {VELOCIDAD * len(list(SPRINTS))} puntos)")

if sobrecargados:
    print(f"\nAcción de planeación: los sprints {sobrecargados} superan la capacidad estimada.")
    print("Decisión: mover historias de prioridad Medium a sprints con holgura antes de iniciar el sprint.")
````

## 9. Presupuesto y viabilidad económica

Costos estimados para las 12 semanas (3 meses) de ejecución, más la operación posterior del asistente.


````python
# Presupuesto del proyecto y viabilidad económica

MESES = 3

# (rol, dedicación, costo mensual COP)
EQUIPO = [
    ("Gerente de proyecto",        0.5, 9_000_000),
    ("Product Owner",              0.5, 7_000_000),
    ("Scrum Master",               0.5, 6_000_000),
    ("Desarrollador senior",       1.0, 8_500_000),
    ("Desarrollador",              1.0, 6_500_000),
    ("Ingeniero de datos / IA",    1.0, 9_000_000),
    ("Analista de calidad (QA)",   0.5, 5_500_000),
    ("Diseñador UX",               0.5, 5_000_000),
]

OTROS = [
    ("Servicios de IA (tokens) y nube", 4_000_000),
    ("Licencias y herramientas",        1_500_000),
    ("Reserva de contingencia (10 %)",  0),  # se calcula abajo
]

costo_equipo = sum(ded * costo * MESES for _, ded, costo in EQUIPO)
costo_otros = sum(c for _, c in OTROS) * MESES
subtotal = costo_equipo + costo_otros
contingencia = subtotal * 0.10
total = subtotal + contingencia

print("COSTOS DE PERSONAL (3 meses)")
print(f"{'Rol':<30}{'Dedic.':>8}{'Mensual':>16}{'Total':>18}")
print("-" * 72)
for rol, ded, costo in EQUIPO:
    print(f"{rol:<30}{ded:>8.0%}{costo:>16,.0f}{ded * costo * MESES:>18,.0f}")
print("-" * 72)
print(f"{'Subtotal personal':<54}{costo_equipo:>18,.0f}")

print("\nOTROS COSTOS (3 meses)")
for concepto, valor in OTROS[:-1]:
    print(f"{concepto:<54}{valor * MESES:>18,.0f}")
print(f"{'Reserva de contingencia (10 %)':<54}{contingencia:>18,.0f}")
print("-" * 72)
print(f"{'PRESUPUESTO TOTAL DEL PROYECTO':<54}{total:>18,.0f}")

# ---- Viabilidad económica ----
print("\n\nVIABILIDAD ECONÓMICA (horizonte de 24 meses)")

consultas_mes = 6_000
costo_actual = 4_200           # COP por consulta atendida por un agente
costo_objetivo = 1_900         # COP por consulta resuelta por el asistente
automatizacion = 0.55          # meta del 55 % de consultas automatizadas
agentes_evitados = 2           # contrataciones que ya no son necesarias (alternativa B)
costo_agente = 5_000_000       # costo mensual de un agente con prestaciones
operacion_mensual = 1_800_000  # API de IA, nube y soporte del asistente

ahorro_atencion = consultas_mes * automatizacion * (costo_actual - costo_objetivo)
ahorro_contratacion = agentes_evitados * costo_agente
ahorro_mensual = ahorro_atencion + ahorro_contratacion - operacion_mensual
flujos = [-total] + [ahorro_mensual] * 24
tasa_mensual = 0.01            # 1 % mensual (~12,7 % efectivo anual)


def vpn(tasa, flujos):
    return sum(f / (1 + tasa) ** t for t, f in enumerate(flujos))


def payback(flujos):
    acumulado = 0
    for t, f in enumerate(flujos):
        anterior = acumulado
        acumulado += f
        if t > 0 and acumulado >= 0:
            return t - 1 + (-anterior / f)
    return None


roi = (sum(flujos[1:]) - total) / total
meses_payback = payback(flujos)

print(f"Consultas automatizadas por mes          : {consultas_mes * automatizacion:,.0f}")
print(f"Ahorro por menor costo de atención       : ${ahorro_atencion:,.0f}")
print(f"Ahorro por contrataciones evitadas       : ${ahorro_contratacion:,.0f}")
print(f"Costo de operación del asistente         : ${operacion_mensual:,.0f}")
print(f"Ahorro neto mensual                      : ${ahorro_mensual:,.0f}")
print(f"Inversión inicial                        : ${total:,.0f}")
print(f"VPN a 24 meses (1 % mensual)             : ${vpn(tasa_mensual, flujos):,.0f}")
print(f"ROI a 24 meses                           : {roi:.1%}")
if meses_payback is None:
    print("Periodo de recuperación                  : no se recupera en 24 meses")
    print("\nDecisión: revisar supuestos; el proyecto no es viable con estos beneficios.")
else:
    print(f"Periodo de recuperación                  : {meses_payback:.1f} meses")
    print(f"\nDecisión: VPN positivo y recuperación en {meses_payback:.0f} meses → el proyecto es viable.")
````

## 10. Mapeo PMBOK ↔ Jira

Esta es la parte central del trabajo: **cómo se traduce cada concepto de gerencia a la herramienta**.

| Concepto de gerencia | Dónde se implementa en Jira |
|---|---|
| Acta constitutiva | Descripción del proyecto + página de Confluence enlazada (o incidencia fijada en el backlog) |
| EDT / paquetes de trabajo | **Épicas** |
| Paquetes de trabajo detallados | **Historias** y **subtareas** |
| Cronograma | **Timeline / Roadmap** con las épicas fechadas |
| Línea base de alcance | Backlog al cierre del sprint 1, con versión etiquetada |
| Estimación | Campo **Story points** |
| Asignación de responsabilidades (RACI) | **Assignee** = A/R · **Reporter** y menciones = C · **Watchers** = I |
| Control de cambios | Tipo de incidencia **Solicitud de cambio** con flujo de aprobación |
| Gestión de riesgos | Tipo de incidencia **Riesgo** con campos de probabilidad e impacto (etiquetas) |
| Políticas y DoD | Descripción del proyecto y checklist en la plantilla de historia |
| Límites y escalamiento | **Prioridades** + regla de **automatización** que notifica al superar el umbral |
| Gobernanza del flujo | **Workflow** con estado de revisión obligatorio antes de *Listo* |
| Ritmo de gobierno | **Sprints** de 2 semanas y reportes por sprint |
| Métricas / gestión en cifras | **Burndown**, **Velocity** y **Cumulative Flow** en Reports y Dashboard |
| Cierre | Versión (*release*) marcada como liberada + informe final |


````python
# Validación de la matriz RACI (regla: una sola A por actividad)

ROLES = ["Patroc.", "PM", "PO", "SM", "Dev", "TI", "Legal"]

RACI = {
    "Aprobación del acta constitutiva":        ["A", "R", "C", "I", "I", "C", "I"],
    "Priorización del backlog":                ["I", "C", "A/R", "I", "C", "I", "I"],
    "Definición de la arquitectura":           ["I", "C", "I", "I", "A/R", "C", "I"],
    "Curaduría del corpus documental":         ["I", "C", "A", "I", "R", "I", "C"],
    "Desarrollo del asistente":                ["I", "I", "C", "I", "A/R", "I", "I"],
    "Eliminación de impedimentos":             ["I", "C", "I", "A/R", "C", "C", "I"],
    "Revisión de privacidad y datos":          ["I", "C", "C", "I", "C", "I", "A/R"],
    "Pruebas y aceptación funcional":          ["I", "C", "A", "I", "R", "I", "I"],
    "Despliegue a producción":                 ["A", "R", "C", "I", "R", "C", "I"],
    "Capacitación a los agentes":              ["I", "A/R", "C", "I", "C", "I", "I"],
    "Cierre del proyecto":                     ["A", "R", "C", "C", "I", "I", "I"],
}

errores = 0
for actividad, valores in RACI.items():
    n_a = sum("A" in v for v in valores)
    n_r = sum("R" in v for v in valores)
    problemas = []
    if n_a != 1:
        problemas.append(f"{n_a} responsables (A)")
    if n_r < 1:
        problemas.append("sin encargado (R)")
    if problemas:
        errores += 1
        print(f"  [X] {actividad}: {', '.join(problemas)}")

print(f"Actividades revisadas: {len(RACI)}  |  Incumplimientos: {errores}")
if errores == 0:
    print("La matriz cumple: una sola A y al menos una R por actividad.")

print("\nCarga de rendición de cuentas (A) por rol:")
for i, rol in enumerate(ROLES):
    acts = [a for a, v in RACI.items() if "A" in v[i]]
    print(f"  {rol:<10}{len(acts)}  {', '.join(acts) if acts else '—'}")
````

### 11.1 Generación de los archivos CSV

En lugar de digitar 50 incidencias a mano, el backlog definido en la sección 6 se exporta a CSV y se
importa a Jira. Se generan **dos archivos** porque las épicas deben existir antes de poder enlazar las
historias a ellas.


````python
# Generación de los archivos CSV para importar a Jira

import csv
import os

CARPETA = "jira"
os.makedirs(CARPETA, exist_ok=True)

# --- Archivo 1: épicas ---
ruta_epicas = os.path.join(CARPETA, "1_epicas.csv")
with open(ruta_epicas, "w", newline="", encoding="utf-8-sig") as f:
    w = csv.writer(f)
    w.writerow(["Issue Type", "Summary", "Description", "Labels"])
    for clave, nombre, objetivo in EPICAS:
        w.writerow(["Epic", f"{clave}. {nombre}", objetivo, "asistia"])

# --- Archivo 2: historias ---
# Claves reales de las épicas en Jira. Se completan DESPUÉS de importar 1_epicas.csv,
# porque Jira enlaza la historia con su épica por clave y no por nombre.
CLAVES_EPICAS = {
    "E1": "ASIST-5",   "E2": "ASIST-6",   "E3": "ASIST-7",   "E4": "ASIST-8",
    "E5": "ASIST-9",   "E6": "ASIST-10",  "E7": "ASIST-11",  "E8": "ASIST-12",
    "E9": "ASIST-13",  "E10": "ASIST-14",
}

# El importador de Jira exige una columna de ID de actividad cuando se usa el campo Principal.
ruta_historias = os.path.join(CARPETA, "2_historias.csv")
with open(ruta_historias, "w", newline="", encoding="utf-8-sig") as f:
    w = csv.writer(f)
    w.writerow(["Issue Id", "Issue Type", "Summary", "Description", "Priority",
                "Story Points", "Sprint", "Labels", "Parent", "Epic"])
    for n, (epica, resumen, desc, prioridad, puntos, sprint) in enumerate(BACKLOG, start=101):
        w.writerow([n, "Historia", resumen, desc, prioridad, puntos,
                    f"Sprint {sprint}", f"asistia {epica.lower()}",
                    CLAVES_EPICAS.get(epica, ""),
                    f"{epica}. {nombres_epicas[epica]}"])

# --- Archivo 2b (respaldo): épicas e historias juntas, enlazadas por ID interno ---
# Sirve si el importador no acepta la clave de una épica ya existente en el campo Principal.
# Requiere borrar antes las épicas ya importadas para no duplicarlas.
ruta_combinado = os.path.join(CARPETA, "2b_epicas_e_historias.csv")
ids_epicas = {clave: i for i, (clave, _, _) in enumerate(EPICAS, start=1)}
with open(ruta_combinado, "w", newline="", encoding="utf-8-sig") as f:
    w = csv.writer(f)
    w.writerow(["Issue Id", "Parent Id", "Issue Type", "Summary", "Description",
                "Priority", "Story Points", "Sprint", "Labels"])
    for clave, nombre, objetivo in EPICAS:
        w.writerow([ids_epicas[clave], "", "Epic", f"{clave}. {nombre}", objetivo,
                    "Medium", "", "", "asistia"])
    for n, (epica, resumen, desc, prioridad, puntos, sprint) in enumerate(BACKLOG, start=101):
        w.writerow([n, ids_epicas[epica], "Historia", resumen, desc, prioridad, puntos,
                    f"Sprint {sprint}", f"asistia {epica.lower()}"])

# --- Archivo 3: riesgos como incidencias ---
ruta_riesgos = os.path.join(CARPETA, "3_riesgos.csv")
with open(ruta_riesgos, "w", newline="", encoding="utf-8-sig") as f:
    w = csv.writer(f)
    w.writerow(["Issue Type", "Summary", "Description", "Priority", "Labels"])
    for rid, desc, p, i, resp, dueno in RIESGOS:
        valor, nivel, _ = severidad(p, i)
        prioridad = {"CRÍTICA": "Highest", "Alta": "High",
                     "Media": "Medium", "Baja": "Low"}[nivel]
        w.writerow(["Riesgo", f"[RIESGO {rid}] {desc}",
                    f"Probabilidad: {p}/5 | Impacto: {i}/5 | Severidad: {valor} ({nivel})\n"
                    f"Respuesta: {resp}\nResponsable: {dueno}",
                    prioridad, f"riesgo severidad-{nivel.lower()}"])

for ruta in (ruta_epicas, ruta_historias, ruta_riesgos, ruta_combinado):
    filas = sum(1 for _ in open(ruta, encoding="utf-8-sig")) - 1
    print(f"Generado: {ruta:<24} {filas:>3} filas")

print("\nOrden de importación en Jira: 1_epicas.csv → 2_historias.csv → 3_riesgos.csv")
````

### 11.2 Paso a paso en Jira

> Las rutas del menú corresponden a Jira Cloud. La interfaz cambia con frecuencia, así que los nombres
> pueden variar levemente; la lógica se mantiene.

#### Paso 1 — Crear la cuenta (5 minutos)

1. Entrar a **https://www.atlassian.com/software/jira** y pulsar **Get it free**.
2. Registrarse con correo institucional o cuenta de Google.
3. Definir el nombre del sitio, por ejemplo `gerencia-proyectos-kg.atlassian.net`.
4. Seleccionar **Jira** (plan **Free**: hasta 10 usuarios, suficiente para el trabajo).

#### Paso 2 — Crear el proyecto

1. **Create project** → plantilla **Scrum**.
2. Elegir **Team-managed** (gestionado por el equipo): es más simple de configurar.
3. Nombre: `AsistIA — Asistente de IA para Servicio al Cliente`. Clave: `ASIST`.

#### Paso 3 — Habilitar las funciones de gestión

En **Project settings → Features**, activar:

| Función | Para qué sirve en la entrega |
|---|---|
| Backlog | Ver y ordenar el backlog del producto |
| Sprints | Planificar los 6 sprints |
| Timeline (Roadmap) | Mostrar el cronograma con las épicas |
| Reports | Burndown y velocidad |
| Estimation → Story points | Estimación en puntos |

#### Paso 4 — Crear los tipos de incidencia de gerencia

En **Configuración del espacio → Tipos de actividades → Añadir tipo de actividad**, crear tres tipos
al mismo nivel que *Task* (es decir, hijos de *Epic*):

- **Historia** — las 40 historias de usuario del backlog.
- **Riesgo** — para el registro de riesgos.
- **Solicitud de cambio** — para el control de cambios.

> La plantilla Scrum trae por defecto *Epic*, *Feature*, *Task* y *Subtask*, pero **no trae Historia**,
> por eso hay que crearla. Los nombres deben coincidir exactamente con los de los CSV.

Esto demuestra que la gobernanza del PMBOK quedó implementada *dentro* de la herramienta.

#### Paso 5 — Ajustar el flujo de trabajo

La plantilla Scrum ya incluye el estado **En revisión** entre *En curso* y *Finalizado*. Si no estuviera,
se agrega en **Tipos de actividades → Editar flujo de trabajo**.

Ese estado es el que implementa la política de aprobación: nada se da por terminado sin revisión.

#### Paso 6 — Importar el backlog

1. Ejecutar la celda anterior para generar la carpeta `jira/` con los tres CSV.
2. En Jira: **Project settings → Import issues** (o el ícono ⚙ → *System* → *External System Import* → *CSV*).
3. Importar **`1_epicas.csv`** y mapear: `Issue Type` → Tipo, `Summary` → Resumen, `Description` → Descripción.
4. Anotar las **claves** que Jira asignó a las épicas (se ven en la pestaña **Cronograma**) y escribirlas en
   el diccionario `CLAVES_EPICAS` de la celda anterior. Volver a ejecutar la celda: el archivo
   `2_historias.csv` se regenera con la columna **Parent** ya resuelta.
5. Importar **`2_historias.csv`** y mapear: `Story Points` → *Story point estimate*, `Sprint` → *Sprint*,
   `Labels` → *Etiquetas* y `Parent` → *Principal* (el campo del padre). La columna `Epic` se puede ignorar:
   queda solo como referencia legible.
6. Importar **`3_riesgos.csv`**: ya trae el tipo **Riesgo**, solo hay que mapear los campos.

> **Por qué dos pasos:** Jira enlaza cada historia con su épica mediante la **clave** (`ASIST-5`), que solo
> existe después de importar las épicas. Por eso primero se importan las épicas, luego se anotan sus claves
> y al final se importan las historias ya enlazadas.

> **Nota del importador:** al mapear el campo **Principal**, Jira exige que el CSV incluya también una
> columna de **ID de actividad** (`Issue Id`). Por eso `2_historias.csv` la trae con un identificador propio
> para cada historia.

**Plan B — `2b_epicas_e_historias.csv`.** Si el importador no acepta la clave de una épica ya creada en el
campo Principal, este archivo trae **épicas e historias juntas**, enlazadas entre sí por `Issue Id` y
`Parent Id` (identificadores internos del archivo, no claves de Jira). Es el mecanismo nativo del
importador para crear jerarquía en una sola pasada. Para usarlo hay que **borrar antes las épicas ya
importadas**, de lo contrario quedan duplicadas.

**Plan C — enlazar en bloque.** Importar las historias sin el campo Principal y luego, en el Backlog,
seleccionar las 4 historias de cada épica y asignarles el padre en una sola acción. Son 10 operaciones.

2. Arrastrar las historias a cada sprint según la columna `Sprint` del CSV.
3. En el sprint 1: **Edit sprint** → fechas de 2 semanas → objetivo del sprint:

4. Pulsar **Start sprint** en el sprint 1.

#### Paso 8 — Configurar el Timeline (cronograma)

En **Timeline**, asignar a cada épica sus fechas de inicio y fin según la sección 7. Se obtiene el

## 13. Conclusiones

1. **La planeación y la herramienta cumplen funciones distintas.** El caso de negocio, el acta constitutiva
   y el plan de gobernanza justifican y autorizan el proyecto; Jira gestiona la ejecución. Ninguno reemplaza
   al otro.
2. **La EDT se traduce naturalmente a épicas e historias**, lo que permite mantener trazabilidad entre el
   alcance planeado y el trabajo ejecutado.
3. **La gobernanza se puede configurar, no solo documentar:** tipos de incidencia para riesgos y cambios,
   un estado de revisión obligatorio y una regla de automatización que aplica el escalamiento.
4. **Un proyecto de IA es mayoritariamente no técnico:** de las 10 épicas, solo tres son de construcción.
   Datos, cumplimiento, adopción y gobernanza pesan más que el modelo en sí.
5. **El valor se mide en el negocio, no en el entregable:** el éxito no es "el asistente funciona", sino
   el 55 % de consultas automatizadas y la recuperación de la inversión, en línea con la idea de valor por
   unidad de inversión.

### Limitaciones del ejercicio

- Las cifras de volumen de consultas, costos y salarios son **supuestos razonables para el ejercicio**, no
  datos de una empresa real.
- La velocidad de 32 puntos por sprint es una estimación inicial: en un proyecto real se ajusta con la
  velocidad observada en los primeros dos sprints.
