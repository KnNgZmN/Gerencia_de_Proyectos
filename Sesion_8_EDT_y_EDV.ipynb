## 1. El desafío del alcance: superando el caos

### 1.1 El problema

Un proyecto enunciado como un todo —*"desarrollar el sistema"*— es **imposible de estimar, asignar y
controlar**. Sin descomposición aparecen los síntomas típicos:

| Síntoma | Qué lo causa |
|---|---|
| No se sabe cuánto falta | No hay unidades de trabajo medibles |
| Aparece trabajo no previsto (*scope creep*) | El alcance nunca se delimitó por escrito |
| Nadie se siente dueño de nada | El trabajo no tiene responsable asignado |
| Las estimaciones fallan | Se estima sobre bloques demasiado grandes |
| Se entrega tarde y sin valor | Se ejecutó sin conexión con el objetivo del negocio |

### 1.2 Las dos respuestas

```
                        PROYECTO (caos inicial)
                                 │
              ┌──────────────────┴──────────────────┐
              ▼                                     ▼
   ┌──────────────────────┐              ┌──────────────────────┐
   │   EDT / WBS          │              │   EDV / VBS          │
   │   Work Breakdown     │              │   Value Breakdown    │
   │   Structure          │              │   Structure          │
   ├──────────────────────┤              ├──────────────────────┤
   │ Descompone el        │              │ Descompone el        │
   │ TRABAJO              │              │ VALOR                │
   │                      │              │                      │
   │ Pregunta:            │              │ Pregunta:            │
   │ ¿qué hay que hacer?  │              │ ¿para qué sirve?     │
   │                      │              │                      │
   │ Produce:             │              │ Produce:             │
   │ CONTROL ABSOLUTO     │              │ IMPACTO REAL         │
   └──────────┬───────────┘              └──────────┬───────────┘
              └──────────────────┬──────────────────┘
                                 ▼
                   ┌───────────────────────────┐
                   │  ALINEACIÓN ESTRATÉGICA   │
                   └───────────────────────────┘
```

No compiten entre sí: la EDT da **control** y la EDV da **dirección**. Una EDT sin EDV produce proyectos
ordenados que no le sirven a nadie; una EDV sin EDT produce buenas intenciones que nadie ejecuta.

## 2. EDT clásica (Estructura de Desglose del Trabajo)

### 2.1 Definición

> **La EDT es la descomposición jerárquica del alcance del proyecto para lograr los objetivos y crear los
> entregables requeridos.**

Dos precisiones que suelen preguntarse en evaluación:

- Descompone **entregables**, no actividades. Los verbos vienen después, en el cronograma.
- Es **jerárquica**: cada nivel detalla al nivel superior, sin agregar ni quitar alcance.

### 2.2 La regla de oro: el 100 %

> **Los niveles inferiores de una rama deben sumar exactamente el 100 % del nivel superior.**

| Si suman **menos** del 100 % | Si suman **más** del 100 % |
|---|---|
| Falta alcance: hay trabajo necesario que nadie planeó ni presupuestó | Sobra alcance: se está haciendo trabajo que no se pidió (*gold plating*) |

### 2.3 La analogía de la ciudad

```
   SECTOR DE LA CIUDAD        ← el alcance total del proyecto
        │
        ├── DISTRITO 1        ← grandes componentes del alcance
        │     ├── MANZANA 1.1
        │     │     ├── EDIFICIO 1.1.1   ← paquete de trabajo: se puede
        │     │     └── EDIFICIO 1.1.2      estimar, asignar y controlar
        │     └── MANZANA 1.2
        └── DISTRITO 2
```

Nadie construye "una ciudad": se construye edificio por edificio. La EDT hace exactamente eso con el
alcance del proyecto.

### 2.4 Los niveles de la EDT

| Nivel | Nombre | Qué contiene | Para qué sirve |
|---|---|---|---|
| 1 | **Producto** (*top-level*) | El alcance total del proyecto | Enunciar de qué se trata todo |
| 2 | **Planificación** | Contenido conocido, sin actividades detalladas | Organizar el alcance en grandes bloques |
| 3 | **Cuentas de control** | Puntos de integración donde se mide el desempeño | Comparar lo planeado contra lo real (alcance, costo, tiempo) |
| 4 | **Paquete de trabajo** | El nivel más bajo de la EDT | Estimar costos y recursos, y asignar un responsable |

```
   ┌──────────────┐   ┌──────────────┐   ┌──────────────┐   ┌──────────────┐
   │  PRODUCTO    │ → │ PLANIFICACIÓN│ → │   CUENTAS    │ → │   PAQUETE    │
   │              │   │              │   │  DE CONTROL  │   │  DE TRABAJO  │
   ├──────────────┤   ├──────────────┤   ├──────────────┤   ├──────────────┤
   │ Alcance      │   │ Contenido    │   │ Puntos de    │   │ Nivel más    │
   │ total del    │   │ conocido sin │   │ integración  │   │ bajo: costos │
   │ proyecto     │   │ actividades  │   │ para medir   │   │ y recursos   │
   │              │   │ detalladas   │   │ el desempeño │   │ estimables   │
   └──────────────┘   └──────────────┘   └──────────────┘   └──────────────┘
        nivel 1            nivel 2            nivel 3            nivel 4
```

### 2.5 Reglas prácticas de construcción

1. **Nombrar con sustantivos**, no con verbos: "Manual de usuario", no "Escribir el manual".
2. **Codificación jerárquica**: 1, 1.1, 1.1.1 — permite rastrear cualquier elemento hasta su origen.
3. **Regla 8/80**: un paquete de trabajo razonable toma entre 8 y 80 horas de esfuerzo.
4. **Un solo responsable por paquete**, coherente con la A de la matriz RACI.
5. **Incluir la gestión del proyecto** como rama: también consume esfuerzo y presupuesto.
6. **No detallar de más**: si un paquete ya se puede estimar y asignar, no hace falta partirlo.

### 2.6 Ejemplo de EDT para un producto de software

```
   SISTEMA DE GESTIÓN
   ├── Administración del proyecto
   ├── Requisitos
   ├── Diseño
   ├── Construcción
   ├── Pruebas
   ├── Documentación
   └── Programa de capacitación
```

Nótese que **administración, documentación y capacitación** son ramas de la EDT: si no aparecen, no se
estiman, y son justamente las que suelen desbordar el presupuesto al final del proyecto.

## 3. Diccionario de la EDT

La EDT muestra **qué** hay que entregar; el diccionario define **con qué detalle y bajo qué criterios**.
Es la ficha técnica de cada paquete de trabajo y el documento que evita discusiones al momento de aceptar.

### 3.1 Estructura mínima

| Código | Descripción | Responsable | Criterios de aceptación |
|---|---|---|---|
| 1.1.1 | Base de datos | Backend | 99 % de disponibilidad (*uptime*) |
| 1.1.2 | Interfaz de usuario | Frontend | Cumplir las pruebas de usabilidad |

### 3.2 Campos completos de una ficha

| Campo | Contenido |
|---|---|
| **Código** | Identificador jerárquico (1.1.1) |
| **Nombre del paquete** | Sustantivo que nombra el entregable |
| **Descripción** | Qué incluye y qué no incluye |
| **Responsable** | Rol o persona que responde por el paquete |
| **Criterios de aceptación** | Condición objetiva para darlo por terminado |
| **Esfuerzo y costo estimados** | Base del presupuesto |
| **Dependencias** | Qué se necesita antes de poder empezar |
| **Hitos asociados** | Con qué punto de control se relaciona |

> **Por qué importa:** sin criterios de aceptación escritos, "terminado" es una opinión. El diccionario
> convierte el alcance en algo verificable.

## 4. Desglose de Valor (EDV): el giro moderno

### 4.1 Definición

> **La EDV es la estructura jerárquica que conecta el alcance del proyecto con su valor de negocio esperado.**

### 4.2 El giro argumental

```
        ENFOQUE TRADICIONAL                    ENFOQUE MODERNO
        ───────────────────                    ───────────────
        Triple restricción                     Flujos de valor
        (alcance · tiempo · costo)             (value streams)

        "¿Entregamos a tiempo,                 "¿El cliente está mejor
         en presupuesto y                       que antes gracias a
         con todo el alcance?"                  lo que entregamos?"

        Mide CUMPLIMIENTO                      Mide IMPACTO
```

El proyecto deja de medirse solo por cumplir la triple restricción y pasa a medirse por el **valor que
habilita**: ingresos, ahorro, satisfacción, reducción de riesgo, posicionamiento.

### 4.3 EDT frente a EDV

| | **EDT (WBS)** | **EDV (VBS)** |
|---|---|---|
| Descompone | El trabajo y los entregables | El valor y los beneficios |
| Pregunta | ¿Qué hay que construir? | ¿Para qué sirve lo que construimos? |
| Unidad mínima | Paquete de trabajo | Beneficio medible |
| Responde ante | El cronograma y el presupuesto | El caso de negocio |
| Aporta | **Control absoluto** | **Impacto real** |
| Riesgo si falta | Caos en la ejecución | Entregar cosas que nadie necesitaba |

### 4.4 Cómo se conectan

```
   EDV (de arriba hacia abajo: por qué)      EDT (de abajo hacia arriba: qué)

   Objetivo estratégico                      Producto
        │                                        │
   Flujo de valor                            Planificación
        │                                        │
   Beneficio medible          ◄──── enlace ──►  Cuenta de control
        │                                        │
   Métrica de valor                          Paquete de trabajo
```

El punto de encuentro es la pregunta: **¿a qué beneficio contribuye este paquete de trabajo?** Si algún
paquete no responde, es candidato a eliminarse; si algún beneficio no tiene paquetes, está sin ejecutar.

## 5. Tutorial práctico: software modular

Se construye la EDT de un **sistema de gestión modular** y luego su EDV. Las celdas siguientes permiten
recorrer la estructura, validar la regla del 100 % y generar el diccionario.

### 5.1 Los seis pasos

| Paso | Acción | Resultado |
|---|---|---|
| 1 | Definir el producto (nivel 1) | Enunciado del alcance total |
| 2 | Identificar los grandes componentes (nivel 2) | Ramas de planificación |
| 3 | Establecer cuentas de control (nivel 3) | Puntos donde se medirá el desempeño |
| 4 | Descomponer en paquetes de trabajo (nivel 4) | Unidades estimables y asignables |
| 5 | Verificar la regla del 100 % | EDT consistente, sin faltantes ni excesos |
| 6 | Escribir el diccionario | Criterios de aceptación por paquete |


````python
# EDT del sistema de gestión modular
# (código, nombre, peso % dentro de su rama padre)

EDT = [
    ("1",     "Sistema de Gestión Modular",        100),

    ("1.1",   "Administración del proyecto",        15),
    ("1.1.1", "Planificación",                      40),
    ("1.1.2", "Seguimiento y control",               40),
    ("1.1.3", "Cierre",                              20),

    ("1.2",   "Módulo de base de datos",             20),
    ("1.2.1", "Modelo de datos",                     40),
    ("1.2.2", "Implementación de la base de datos",  40),
    ("1.2.3", "Migración de datos",                  20),

    ("1.3",   "Módulo de servicios (backend)",       25),
    ("1.3.1", "Servicios de autenticación",          30),
    ("1.3.2", "Servicios de negocio",                50),
    ("1.3.3", "Integraciones externas",              20),

    ("1.4",   "Módulo de interfaz (frontend)",       20),
    ("1.4.1", "Diseño de la interfaz",               30),
    ("1.4.2", "Pantallas principales",               50),
    ("1.4.3", "Accesibilidad",                       20),

    ("1.5",   "Calidad y pruebas",                   10),
    ("1.5.1", "Plan de pruebas",                     30),
    ("1.5.2", "Pruebas funcionales",                 40),
    ("1.5.3", "Pruebas de desempeño",                30),

    ("1.6",   "Documentación y capacitación",        10),
    ("1.6.1", "Manual de usuario",                   40),
    ("1.6.2", "Manual técnico",                      30),
    ("1.6.3", "Programa de capacitación",            30),
]

# Impresión jerárquica
for codigo, nombre, peso in EDT:
    nivel = codigo.count(".")
    sangria = "    " * nivel
    marca = "[PT]" if nivel == 2 else "    "
    print(f"{marca} {codigo:<8}{sangria}{nombre}  ({peso} %)")

paquetes = [e for e in EDT if e[0].count(".") == 2]
print(f"\nNiveles: 3  |  Cuentas de control: {len([e for e in EDT if e[0].count('.') == 1])}"
      f"  |  Paquetes de trabajo: {len(paquetes)}")
````

````python
# Validación de la regla del 100 %

from collections import defaultdict

hijos = defaultdict(list)
for codigo, nombre, peso in EDT:
    if "." in codigo:
        padre = codigo.rsplit(".", 1)[0]
        hijos[padre].append((codigo, nombre, peso))

nombres = {c: n for c, n, _ in EDT}
errores = 0

print(f"{'Rama':<8}{'Nombre':<36}{'Suma':>6}   Estado")
print("-" * 70)
for padre in sorted(hijos, key=lambda c: [int(x) for x in c.split(".")]):
    suma = sum(p for _, _, p in hijos[padre])
    if suma == 100:
        estado = "OK"
    elif suma < 100:
        estado = f"FALTA ALCANCE ({100 - suma} % sin planear)"
        errores += 1
    else:
        estado = f"EXCESO DE ALCANCE ({suma - 100} % de más)"
        errores += 1
    print(f"{padre:<8}{nombres[padre]:<36}{suma:>5} %   {estado}")

print("-" * 70)
print("La EDT cumple la regla del 100 %." if errores == 0
      else f"Hay {errores} rama(s) que incumplen la regla del 100 %.")

# Peso absoluto de cada paquete dentro del proyecto completo
print("\nPeso de cada paquete de trabajo sobre el proyecto total:")
for codigo, nombre, peso in EDT:
    if codigo.count(".") == 2:
        padre = codigo.rsplit(".", 1)[0]
        peso_padre = next(p for c, _, p in EDT if c == padre)
        absoluto = peso_padre * peso / 100
        print(f"  {codigo:<8}{nombre:<36}{absoluto:>5.1f} %  {'#' * round(absoluto)}")
````

````python
# Diccionario de la EDT

# código: (responsable, criterios de aceptación, esfuerzo en horas)
DICCIONARIO = {
    "1.2.1": ("Arquitecto de datos", "Modelo normalizado y aprobado por el PO", 40),
    "1.2.2": ("Backend", "99 % de disponibilidad en pruebas de 72 horas", 60),
    "1.2.3": ("Backend", "Cero registros perdidos en la migración de prueba", 32),
    "1.3.1": ("Backend", "Autenticación con doble factor funcionando", 48),
    "1.3.2": ("Backend", "Todos los casos de uso del sprint aprobados", 80),
    "1.3.3": ("Backend", "Integración probada con el sistema contable", 40),
    "1.4.1": ("Diseño UX", "Prototipo validado con 5 usuarios reales", 32),
    "1.4.2": ("Frontend", "Pruebas de usabilidad superadas", 72),
    "1.4.3": ("Frontend", "Cumplimiento de pautas WCAG nivel AA", 24),
}

print(f"{'Código':<9}{'Descripción':<36}{'Responsable':<22}{'Horas':>6}  Criterios")
print("-" * 130)
for codigo, nombre, _ in EDT:
    if codigo in DICCIONARIO:
        responsable, criterio, horas = DICCIONARIO[codigo]
        alerta = "" if 8 <= horas <= 80 else "  <- fuera de la regla 8/80"
        print(f"{codigo:<9}{nombre:<36}{responsable:<22}{horas:>6}  {criterio}{alerta}")

total_horas = sum(h for _, _, h in DICCIONARIO.values())
print("-" * 130)
print(f"Paquetes documentados: {len(DICCIONARIO)} de {len(paquetes)}  |  Esfuerzo documentado: {total_horas} horas")
print("Regla 8/80: ningún paquete debe estar por debajo de 8 ni por encima de 80 horas.")
````

````python
# EDV: del paquete de trabajo al valor de negocio

# (rama de la EDT, flujo de valor, beneficio esperado, métrica, meta)
EDV = [
    ("1.2 Base de datos",     "Confiabilidad de la información",
     "Menos errores por datos inconsistentes", "Incidencias por datos/mes", "de 45 a 5"),
    ("1.3 Servicios",         "Automatización de la operación",
     "Menor tiempo de procesamiento de pedidos", "Minutos por pedido", "de 12 a 3"),
    ("1.4 Interfaz",          "Experiencia de usuario",
     "Mayor adopción de la herramienta", "Usuarios activos/semana", "de 60 a 180"),
    ("1.5 Calidad",           "Reducción de riesgo",
     "Menos defectos en producción", "Defectos críticos/trimestre", "de 8 a 1"),
    ("1.6 Capacitación",      "Adopción organizacional",
     "Autonomía de los usuarios", "Tickets de soporte/mes", "de 120 a 40"),
    ("1.1 Administración",    "Gobernanza del proyecto",
     "Decisiones oportunas y trazables", "Desviación de cronograma", "menor al 10 %"),
]

print(f"{'Rama de la EDT':<22}{'Flujo de valor':<34}{'Métrica':<30}Meta")
print("-" * 110)
for rama, flujo, beneficio, metrica, meta in EDV:
    print(f"{rama:<22}{flujo:<34}{metrica:<30}{meta}")

print("\nBeneficio esperado por rama:")
for rama, _, beneficio, _, _ in EDV:
    print(f"  {rama:<22}->  {beneficio}")

# Verificación cruzada: toda rama de la EDT debe aportar a algún flujo de valor
ramas_edt = {c for c, _, _ in EDT if c.count(".") == 1}
ramas_edv = {r.split()[0] for r, _, _, _, _ in EDV}
huerfanas = ramas_edt - ramas_edv
print(f"\nRamas de la EDT sin valor asociado: {sorted(huerfanas) if huerfanas else 'ninguna'}")
print("Toda rama de trabajo debe poder responder: ¿a qué beneficio del negocio contribuye?")
````

## 6. Conclusiones

1. **La EDT transforma el caos en pasos manejables**: descompone el alcance en entregables hasta llegar a
   paquetes de trabajo que se pueden estimar, asignar y controlar.
2. **La regla del 100 %** es el criterio de consistencia: si los hijos no suman el total del padre, o falta
   alcance o sobra trabajo.
3. **Los cuatro niveles** tienen funciones distintas: producto, planificación, cuentas de control y paquete
   de trabajo. Las cuentas de control son el punto donde realmente se mide el desempeño.
4. **El diccionario convierte el alcance en algo verificable**, porque escribe los criterios de aceptación
   antes de empezar a trabajar.
5. **La EDV es el giro moderno**: pasar de medir cumplimiento de la triple restricción a medir el impacto
   en el negocio, conectando cada rama del trabajo con un beneficio medible.
6. **No son excluyentes.** EDT sin EDV da control sobre trabajo que quizá no sirve; EDV sin EDT da
   intenciones sin ejecución. Juntas producen **alineación estratégica**.
