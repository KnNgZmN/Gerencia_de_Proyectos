## 2. Marco doctrinal y estratégico

### 2.1 Los siete Principios Fundamentales

Todo proyecto de la Cruz Roja Colombiana debe ser coherente con los principios del Movimiento
Internacional de la Cruz Roja y de la Media Luna Roja. Funcionan como **restricción de diseño**: un
proyecto que comprometa la neutralidad o la imparcialidad no es viable, sin importar su financiación.

| Principio | Implicación para la gestión de proyectos |
|---|---|
| **Humanidad** | El fin del proyecto es prevenir y aliviar el sufrimiento humano |
| **Imparcialidad** | La población beneficiaria se prioriza solo por necesidad, sin discriminación |
| **Neutralidad** | No se toma partido en hostilidades ni en controversias políticas o ideológicas |
| **Independencia** | Los aliados y donantes no pueden condicionar la acción humanitaria |
| **Voluntariado** | La ejecución se apoya en voluntarios, sin ánimo de lucro |
| **Unidad** | Una sola Sociedad Nacional en el país, abierta a todos |
| **Universalidad** | Cooperación con las demás Sociedades Nacionales del Movimiento |

Estos principios se complementan con los marcos de **Derechos Humanos (DDHH)** y **Derecho Internacional
Humanitario (DIH)**, que son componente explícito de la línea de Construcción de Paz y Doctrina.

### 2.2 Plan Estratégico Nacional 2021–2025

El plan quinquenal fija los propósitos de la organización. Según lo publicado, su visión es que la Cruz Roja
Colombiana sea la **institución humanitaria de referencia en el país** para el desarrollo social y humanitario
y la **resiliencia de las comunidades vulnerables**, apoyada en:

- un **voluntariado sólido**,
- **aliados estratégicos** que fortalezcan la misión,
- **sostenibilidad integral**, y
- **procesos eficientes** orientados a la innovación humanitaria.

Dos orientaciones transversales atraviesan todas las líneas: la **transformación de comunidades vulnerables**
con acciones integradas de desarrollo social y humanitario, y la **mitigación y adaptación al cambio
climático** incorporadas de forma integral en la acción humanitaria.

## 3. Las cinco líneas programáticas

Las líneas programáticas son la forma en que la organización **agrupa sus proyectos por propósito**.
En términos de gerencia de proyectos, cumplen el papel de **programas dentro de un portafolio**: cada
una reúne proyectos relacionados que persiguen un mismo beneficio estratégico.

| Línea programática | Propósito | Poblaciones prioritarias |
|---|---|---|
| Gestión del Riesgo de Desastres | Seguridad, bienestar y desarrollo sostenible frente a amenazas | Comunidades vulnerables y afectadas por desastres |
| Gestión Integral de Salud | Mejorar la salud y prevenir enfermedades | Población vulnerable no asegurada, víctimas, personal sanitario |
| Construcción de Paz y Doctrina Institucional | Prevenir y mitigar el impacto humanitario del conflicto y la violencia | Víctimas, jóvenes, comunidades educativas |
| Educación | Formar capacidades humanitarias | Voluntarios, personal y comunidades |
| Desarrollo de Seccionales | Fortalecer las capacidades de respuesta local | Las 31 seccionales del país |


### 3.1 Gestión del Riesgo de Desastres

> **Objetivo:** contribuir a la seguridad, el bienestar, la calidad de vida de las personas y el desarrollo
> sostenible mediante procesos integrales de gestión del riesgo.

```
   GESTIÓN DEL RIESGO DE DESASTRES
   │
   ├── 1. Conocimiento del riesgo
   │      ├── Identificación de escenarios de amenaza
   │      ├── Análisis y monitoreo del riesgo
   │      └── Comunicación y sensibilización
   │
   ├── 2. Reducción del riesgo
   │      ├── Medidas de mitigación y prevención
   │      └── Disminución de la vulnerabilidad
   │
   └── 3. Manejo de desastres
          ├── Preparación
          ├── Respuesta: búsqueda y rescate, asistencia, alojamiento temporal, agua y saneamiento
          └── Recuperación posdesastre
```

**Líneas de acción:** incidencia en todos los niveles de gobierno, formación y fortalecimiento de
capacidades, innovación tecnológica, infraestructura, protección de medios de vida, preparación ante
desastres y **adaptación al cambio climático**.

**Enfoque de intervención:** integral, con **empoderamiento comunitario y planeación participativa**,
coordinado con los consejos territoriales de gestión del riesgo y con aliados internacionales.

**Proyecto de referencia:** *Ciudades por la Acción Climática* — fortalece la resiliencia urbana en
**Neiva y Sincelejo** mediante acciones locales.


### 3.2 Gestión Integral de Salud

> **Objetivo:** mejorar la salud y prevenir enfermedades en las comunidades más vulnerables mediante
> atención primaria con altos estándares de calidad, bajo los principios de imparcialidad, neutralidad y humanidad.

```
   GESTIÓN INTEGRAL DE SALUD
   │
   ├── Prestación de Servicios de Salud
   │      └── Acceso, calidad y eficiencia; redes integradas y capacidades territoriales
   │
   ├── Misión Médica
   │      └── Protección del personal sanitario frente a agresiones en zonas de conflicto
   │
   ├── Salud Mental y Apoyo Psicosocial
   │      └── Intervenciones con voluntarios y profesionales para comunidades resilientes
   │
   └── Salud Pública en Emergencias
          └── Gestión de riesgos sanitarios por desastres naturales y situaciones antrópicas
```

**Poblaciones:** comunidades vulnerables y pobres no aseguradas, personas afectadas por emergencias,
víctimas del conflicto armado y personal sanitario en riesgo.

**Enfoque:** prevención, calidad técnica, participación comunitaria y articulación intersectorial.


### 3.3 Construcción de Paz y Doctrina Institucional

> **Objetivo:** prevenir y mitigar el impacto humanitario del conflicto armado y la violencia, atendiendo
> las causas de vulnerabilidad y aumentando la capacidad institucional de respuesta en contextos sensibles.

```
   CONSTRUCCIÓN DE PAZ Y DOCTRINA INSTITUCIONAL
   │
   ├── Doctrina Institucional
   │      ├── Acceso más seguro: preparar a la institución para actuar en contextos de inseguridad
   │      ├── Comunicación operacional: difundir la acción humanitaria y proteger a sus integrantes
   │      └── DDHH y DIH: sensibilización y formación en sistemas de protección
   │
   └── Prevención de la Violencia
          ├── Educación, promoción de derechos y construcción de paz en violencias urbanas
          ├── Enfoque de género y diversidad
          └── Mitigación de consecuencias humanitarias
```

**Poblaciones:** voluntarios y empleados, jóvenes del Sistema de Responsabilidad Penal, comunidades
educativas, víctimas del conflicto armado y la violencia urbana, y redes comunitarias.

**Particularidad para la gestión:** esta línea tiene un papel doble. Además de ejecutar proyectos propios,
produce las **condiciones de seguridad y aceptación** (acceso más seguro, comunicación operacional) que
permiten que los proyectos de las demás líneas puedan operar en zonas de conflicto.


### 3.4 Educación

Línea orientada a **formar capacidades humanitarias** en el voluntariado, el personal institucional y las
comunidades. En la lógica del portafolio actúa como **habilitadora**: los proyectos de riesgo, salud y paz
dependen de personas capacitadas para ejecutarlos con calidad técnica y apego a los principios.

### 3.5 Desarrollo de Seccionales

Línea enfocada en el **fortalecimiento misional** de las seccionales: identificar sus capacidades de
intervención y fortalecerlas para que respondan de forma oportuna a las necesidades humanitarias de las
comunidades locales. En la lógica del portafolio es la línea que **construye la capacidad de ejecución**:
sin seccionales fuertes, los proyectos no llegan al territorio.

## 4. Estructura de ejecución

| Actor | Papel en la gestión de proyectos |
|---|---|
| **Sede Nacional** | Define la estrategia, las líneas programáticas y los lineamientos técnicos; consolida la rendición de cuentas |
| **31 Seccionales** | Ejecutan los proyectos en el territorio; conocen las necesidades locales y a las comunidades |
| **Voluntariado (+18.000)** | Capacidad operativa de respuesta y de trabajo comunitario |
| **Sociedades Nacionales participantes y Movimiento Internacional** | Cooperación técnica y financiera entre Sociedades de la Cruz Roja |
| **Estado** | Articulación con consejos territoriales de gestión del riesgo, sistema de salud y autoridades |
| **Aliados y donantes** | Financian proyectos sin condicionar la independencia de la acción humanitaria |
| **Comunidades** | Participan en la planeación y son protagonistas de su propia resiliencia |

La cobertura territorial reportada en 2025 es del **97 %**, lo que solo es posible por el modelo
descentralizado: la estrategia se define en el nivel nacional y la **ejecución ocurre en las seccionales**.

## 5. Ciclo de gestión de un proyecto humanitario

A partir de lo que la organización describe en sus líneas programáticas —diagnóstico de escenarios,
planeación participativa, articulación con aliados, ejecución territorial y reporte de cifras— el ciclo
de un proyecto puede representarse así:

```
   ┌───────────────────┐    ┌───────────────────┐    ┌───────────────────┐
   │ 1. IDENTIFICACIÓN │ →  │ 2. FORMULACIÓN    │ →  │ 3. MOVILIZACIÓN   │
   │   Necesidad       │    │   Objetivos,      │    │   DE RECURSOS     │
   │   humanitaria y   │    │   población,      │    │   Aliados,        │
   │   escenarios de   │    │   línea progra-   │    │   donantes,       │
   │   riesgo          │    │   mática, zona    │    │   cooperación     │
   └───────────────────┘    └───────────────────┘    └─────────┬─────────┘
             ▲                                                 │
             │                                                 ▼
   ┌─────────┴─────────┐    ┌───────────────────┐    ┌───────────────────┐
   │ 6. APRENDIZAJE Y  │ ←  │ 5. SEGUIMIENTO Y  │ ←  │ 4. EJECUCIÓN      │
   │   RENDICIÓN DE    │    │   MEDICIÓN        │    │   Seccionales +   │
   │   CUENTAS         │    │   Indicadores de  │    │   voluntariado +  │
```

La sección *Nuestra gestión en cifras* organiza sus indicadores por población atendida:

Esto muestra que la organización **mide el éxito por población alcanzada y servicios entregados**, no solo

````python
# Distribución de servicios humanitarios 2025 por área (Informe de Gestión 2025)

servicios = {
    "Gestión del riesgo y emergencias": 657_000,
    "Salud": 330_000,
    "Población migrante": 188_000,
    "Construcción de paz": 47_000,
}

total = sum(servicios.values())
escala = 40 / max(servicios.values())

print(f"{'Área':<34}{'Servicios':>10}{'%':>7}")
print("-" * 92)
for area, n in sorted(servicios.items(), key=lambda x: -x[1]):
    barra = "█" * round(n * escala)
    print(f"{area:<34}{n:>10,}{n / total:>7.1%}  {barra}")
print("-" * 92)
print(f"{'Total de las cuatro áreas':<34}{total:>10,}")

inversion = 68_800_000_000
personas = 3_200_000
print(f"\nInversión social por persona beneficiada: ${inversion / personas:,.0f} COP")
````

## 7. Relación con la Guía PMBOK® (Sesión 6)

La estructura de la Cruz Roja Colombiana se corresponde con la jerarquía vista en clase:

```
   PMBOK                                   CRUZ ROJA COLOMBIANA
   ─────                                   ────────────────────
   Estrategia / misión            ←→       Misión + 7 Principios Fundamentales
   Objetivos estratégicos         ←→       Plan Estratégico Nacional 2021–2025
   Portafolio                     ←→       Gestión humanitaria
   Programas                      ←→       5 Líneas programáticas
   Proyectos                      ←→       Proyectos (ej. Ciudades por la Acción Climática)
   Operaciones                    ←→       Servicios permanentes de las seccionales
   Recursos organizacionales      ←→       Seccionales, voluntariado, aliados, inversión social
   Sistema de entrega de valor    ←→       Gestión en cifras / Informe de gestión
```

| Concepto PMBOK | Equivalente en la Cruz Roja | Observación |
|---|---|---|
| Caso de negocio | Diagnóstico de necesidad y propuesta al aliado o donante | Justifica el proyecto por **valor humanitario**, no por rentabilidad |
| Acta constitutiva | Convenio o acuerdo de cooperación | Autoriza, fija alcance, recursos y responsables |
| Objetivos medibles | Metas de beneficiarios y servicios | Base de la *gestión en cifras* |
| Criterios de éxito | Población atendida, cobertura, calidad técnica | Medidos por indicador y rendidos públicamente |
| Valor por unidad de inversión | Inversión social ÷ personas beneficiadas | Aprox. 21.500 COP por persona en 2025 |
| Gestión de interesados | Comunidades, Estado, Movimiento, donantes | La comunidad participa en la planeación |

**Diferencia clave:** en una empresa el caso de negocio se sustenta en retorno financiero; en la Cruz Roja
la viabilidad se evalúa por **necesidad humanitaria, coherencia con los principios y capacidad de ejecución**
en el territorio. El retorno es social y se mide en vidas protegidas y sufrimiento aliviado.
