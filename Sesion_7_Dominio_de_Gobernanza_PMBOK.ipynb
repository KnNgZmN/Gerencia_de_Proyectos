## 1. Marco de Gobernanza

### 1.1 Definición

> **La gobernanza proporciona dirección a través de políticas para cumplir con los objetivos estratégicos.**

Es el **conjunto de reglas y procesos** que determina cómo se toman las decisiones en el proyecto, quién
tiene autoridad para tomarlas y cómo se controla que el trabajo siga **alineado con la estrategia del negocio**.

Funciona como la **estructura de un edificio**: no es el edificio en sí (el producto), pero sin ella nada se
sostiene. Una buena estructura, además, tiene **resiliencia para soportar cambios** sin colapsar.

### 1.2 Características de una buena gobernanza

| Característica | Qué significa en la práctica |
|---|---|
| **Conjunto de reglas y procesos** | Hay políticas escritas, conocidas por todos, no decisiones improvisadas |
| **Alineada con la estrategia del negocio** | Cada decisión del proyecto se puede rastrear hasta un objetivo estratégico |
| **Estructura de integridad** | Transparencia, trazabilidad de decisiones y rendición de cuentas |
| **Resiliencia ante el cambio** | El marco se adapta cuando cambian el contexto, el alcance o las prioridades |

### 1.3 Fundamentos adaptables

La Guía PMBOK® 8 presenta **fundamentos adaptables para la gestión de proyectos**: la gobernanza no es igual
para una *startup*, un portafolio corporativo o un equipo ágil. Se ajusta como una **balanza**:

```
        AGILIDAD (pluma)                                    CONTROL (ancla)
        ────────────────                                    ───────────────
        · Decisiones rápidas                                · Cumplimiento y auditoría
        · Autonomía del equipo           ◄── balance ──►    · Aprobaciones formales
        · Documentación mínima                              · Trazabilidad completa
        · Adecuada para alta                                · Adecuada para alto riesgo
          incertidumbre                                       o regulación
```

Demasiado control **frena** al equipo; demasiada libertad sin límites **dispersa** el esfuerzo y aumenta el
riesgo. El gerente de proyectos adapta (*tailoring*) el nivel de gobernanza al contexto.

### 1.4 Gobernanza vs. gestión

| | Gobernanza | Gestión |
|---|---|---|
| Pregunta | ¿**Quién** decide y **bajo qué reglas**? | ¿**Cómo** se ejecuta el trabajo? |
| Enfoque | Dirección, supervisión, rendición de cuentas | Planificación, ejecución, control diario |
| Horizonte | Estratégico | Operativo y táctico |
| Ejemplo | "Cambios de presupuesto mayores al 10 % los aprueba el patrocinador" | "Reasignar tareas del sprint para cumplir la fecha" |

### 1.5 Los elementos interconectados del desempeño del proyecto

La gobernanza no actúa aislada: el **sistema de toma de decisiones** está en el centro y conecta todos los
elementos del desempeño del proyecto. Una decisión en uno de ellos impacta a los demás.

```
                   Gestión de riesgos ◄──────────► Cronograma
                              ╲                       ╱
                               ╲                     ╱
   Cultura organizacional ◄──── SISTEMA DE TOMA DE ────► Requisitos técnicos
                               ╱   DECISIONES      ╲
                              ╱                     ╲
                   Cultura del equipo ◄──────────► Medición
```

Por ejemplo, decidir acortar el cronograma afecta el riesgo, exige ajustar los requisitos técnicos, pone
presión sobre la cultura del equipo y cambia lo que se debe medir.

## 2. Roles Clave

La gobernanza se hace concreta cuando se define **qué rol decide qué**. En un entorno ágil los roles del
equipo son:

| Rol | Función principal | Qué decide | Qué **no** le corresponde |
|---|---|---|---|
| **Patrocinador** (*Sponsor*) | **Proporciona recursos** | Aprobación final, presupuesto, prioridad estratégica | Decidir cómo se construye el producto |
| **Product Owner** | **Gestiona el backlog** | Qué se construye y en qué orden; acepta o rechaza incrementos | Decidir cómo se implementa técnicamente |
| **Scrum Master** | **Elimina obstáculos** | Cómo mejorar el proceso; facilita y protege al equipo | Asignar tareas o decidir el alcance |
| **Desarrolladores** | **Crean el producto** | Cómo construir: diseño técnico, estimaciones, reparto del trabajo | Cambiar prioridades del backlog |

```
   PATROCINADOR ──── recursos y aprobación ────► PRODUCT OWNER ──── prioridades ────► DESARROLLADORES
                                                                                        ▲
                                                  SCRUM MASTER ──── elimina obstáculos ─┘
```

**Idea clave:** cada rol tiene un **ámbito de decisión propio**. Cuando un rol invade el ámbito de otro (por
ejemplo, el patrocinador que decide detalles técnicos) la gobernanza empieza a **frenar** al equipo.

## 3. Toma de Decisiones

Existen dos modelos extremos de autoridad. La mayoría de los proyectos combinan ambos.

```
        MODELO CENTRALIZADO                          MODELO DISTRIBUIDO
        (pirámide)                                   (red)

               ●                                     o───o───o
              ╱ ╲                                    │ ╲ │ ╱ │
             ╱   ╲                                   o───O───o
            ╱─────╲                                  │ ╱ │ ╲ │
           ╱       ╲                                 o───o───o
          ╱─────────╲

   Liderazgo descendente y un                 Autoridad distribuida, gran
   único punto de responsabilidad.            trabajo en equipo y decisiones locales.
```

| Aspecto | Liderazgo descendente | Autoridad distribuida |
|---|---|---|
| Quién decide | Una persona o nivel jerárquico | Quien tiene el conocimiento, cerca del trabajo |
| Responsabilidad | Un **único punto** de responsabilidad | Compartida por el equipo |
| Velocidad | Lenta si todo sube a la cima | Rápida: se decide localmente |
| Control y trazabilidad | Alto | Requiere acuerdos y transparencia |
| Riesgo principal | Cuello de botella y dependencia del líder | Decisiones inconsistentes si no hay límites |
| Adecuado para | Proyectos predictivos, regulados, de alto riesgo | Proyectos ágiles, alta incertidumbre, innovación |

### ¿Cómo elegir el nivel de decisión?

| Criterio | Decisión local (equipo) | Decisión escalada (PMO / patrocinador) |
|---|---|---|
| Impacto | Afecta solo al trabajo del equipo | Afecta presupuesto, alcance o estrategia |
| Reversibilidad | Fácil de revertir | Difícil o costosa de revertir |
| Conocimiento | Lo tiene el equipo | Requiere visión de negocio o de portafolio |
| Riesgo | Dentro de los umbrales acordados | Supera los umbrales definidos |

## 4. Autogobierno Ágil

> **Los equipos ágiles con límites definidos usan reuniones diarias para autoorganizarse.**

El autogobierno **no es ausencia de gobernanza**: es gobernanza distribuida. La organización fija el
**marco** (objetivos, límites, políticas) y dentro de ese marco el equipo decide cómo trabajar.

```
   ┌───────────────────────────────────────────────────────────┐
   │  LÍMITES DEFINIDOS POR LA GOBERNANZA                      │
   │  objetivo del producto · presupuesto · políticas · fechas │
   │                                                           │
   │     ┌─────────────────────────────────────────────┐       │
   │     │  ESPACIO DE AUTOORGANIZACIÓN DEL EQUIPO     │       │
   │     │  cómo construir · quién toma cada tarea     │       │
   │     │  cómo resolver problemas del día a día      │       │
   │     └─────────────────────────────────────────────┘       │
   │                                                           │
   └───────────────────────────────────────────────────────────┘
```

### Los eventos ágiles como mecanismos de gobernanza

| Evento | Función de gobernanza |
|---|---|
| **Reunión diaria** (*Daily*) | Autoorganización: sincronizar, detectar bloqueos y reajustar el plan del día |
| **Planificación del sprint** | Compromiso del equipo con un objetivo dentro de los límites |
| **Revisión del sprint** | Transparencia ante interesados; el PO acepta o rechaza el incremento |
| **Retrospectiva** | Mejora del propio proceso de trabajo |
| **Backlog priorizado** | Política viva que define qué es lo más valioso |

**Condición para que funcione:** límites claros. Sin límites, la autoorganización se vuelve desorden; con
límites excesivos, se vuelve microgestión.

## 5. Plan de Gobernanza

El plan de gobernanza documenta **cómo se va a gobernar el proyecto**. Se construye en tres pasos
secuenciales:

```
   ┌────────────────────┐      ┌────────────────────┐      ┌────────────────────┐
   │ POLÍTICAS          │      │ LÍMITES            │      │ ESCALAMIENTO       │
   │                    │      │                    │      │                    │
   │ Definir reglas y   │  ->  │ Establecer         │  ->  │ Cuándo involucrar  │
   │ estándares de      │      │ umbrales de        │      │ a la PMO o al      │
   │ cumplimiento.      │      │ riesgo y control.  │      │ Patrocinador.      │
   └────────────────────┘      └────────────────────┘      └────────────────────┘
```

### 5.1 Políticas — definir reglas y estándares de cumplimiento

- Estándares de calidad y definición de terminado.
- Normas de seguridad, legales o regulatorias que aplican.
- Frecuencia y formato de reportes.
- Procedimiento para solicitar cambios.

### 5.2 Límites — establecer umbrales de riesgo y control

Los límites definen **hasta dónde puede decidir cada nivel** sin pedir autorización. Es lo que permite
que el equipo sea autónomo sin perder el control.

| Variable | Decide el equipo / PM | Escala a la PMO | Escala al patrocinador |
|---|---|---|---|
| Desviación de costo | Hasta 5 % | Entre 5 % y 10 % | Más del 10 % |
| Desviación de cronograma | Hasta 1 semana | 1 a 3 semanas | Más de 3 semanas o afecta un hito contractual |
| Cambio de alcance | Ajustes dentro del backlog | Nuevas funcionalidades relevantes | Cambio del objetivo del proyecto |
| Riesgo | Bajo | Medio | Alto o crítico |

> *Los porcentajes son un ejemplo ilustrativo; cada organización define sus propios umbrales.*

### 5.3 Escalamiento — cuándo involucrar a la PMO o al patrocinador

El escalamiento **no es un fracaso** del equipo: es la regla acordada para las decisiones que superan su
límite. Un buen plan de escalamiento define:

1. **Disparador:** qué umbral activa el escalamiento.
2. **Destino:** a quién se escala (PMO, patrocinador, comité).
3. **Tiempo de respuesta:** plazo máximo para decidir, para que el escalamiento no frene el proyecto.
4. **Información mínima:** situación, impacto, alternativas y recomendación.

## 6. Matriz RACI

> **Define quién es Encargado, Responsable, Consultado e Informado en las tareas del proyecto.**

### 6.1 Significado de cada letra

| Letra | Término (inglés) | Término usado en clase | Qué hace |
|---|---|---|---|
| **R** | *Responsible* | **Encargado** | Ejecuta el trabajo |
| **A** | *Accountable* | **Responsable** | Rinde cuentas por el resultado y lo aprueba; tiene la última palabra |
| **C** | *Consulted* | **Consultado** | Aporta su opinión **antes** de decidir (comunicación en dos vías) |
| **I** | *Informed* | **Informado** | Se le comunica el resultado **después** (comunicación en una vía) |

> **Atención:** en español "responsable" suele confundirse. La **R** es quien *hace* el trabajo; la **A** es
> quien *responde* por él.

### 6.2 Matriz vista en clase

| Tarea | Patrocinador | Product Owner | Scrum Master | Desarrolladores |
|---|:---:|:---:|:---:|:---:|
| **Requisitos** | I | **A/R** | C | C |
| **Código** | I | C | I | **A/R** |
| **Bloqueos** | I | I | **A/R*** | C* |
| **Aprobación** | **A/R** | C | I* | I |

\* *Celdas cubiertas por ilustraciones en la diapositiva; el valor se infiere a partir de la función de cada
rol (el Scrum Master elimina obstáculos).*

### 6.3 Lectura de la matriz

- **Requisitos:** los define y responde por ellos el **Product Owner**; consulta al Scrum Master y a los
  desarrolladores (viabilidad técnica) e informa al patrocinador.
- **Código:** lo construyen y responden por él los **desarrolladores**; el PO es consultado para aclarar
  el comportamiento esperado.
- **Bloqueos:** los resuelve el **Scrum Master**, que es quien elimina obstáculos.
- **Aprobación:** la da el **Patrocinador**, que es quien proporciona los recursos; el PO es consultado.

Se observa que la matriz **coincide con los roles de la sección 2**: cada rol es A/R justamente en la tarea
que corresponde a su función principal.

### 6.4 Reglas para construir una buena RACI

1. **Una sola A por tarea.** Si hay dos personas que responden, en la práctica no responde ninguna.
2. **Al menos una R por tarea.** Alguien tiene que ejecutar el trabajo.
3. **Evitar el exceso de C.** Demasiados consultados vuelven lenta la decisión: la gobernanza frena.
4. **Evitar el exceso de A/R en una misma persona.** Indica un cuello de botella.
5. **Cada fila debe ser una tarea o decisión concreta**, no un área difusa.

## 7. Apoyos prácticos

Ejercicios ejecutables de repaso. No requieren librerías externas.


````python
# Validador de una matriz RACI

roles = ["Patrocinador", "Product Owner", "Scrum Master", "Desarrolladores"]

raci = {
    "Requisitos": ["I",   "A/R", "C",   "C"],
    "Código":     ["I",   "C",   "I",   "A/R"],
    "Bloqueos":   ["I",   "I",   "A/R", "C"],
    "Aprobación": ["A/R", "C",   "I",   "I"],
}

# Imprimir la matriz
ancho = 16
print("Tarea".ljust(12) + "".join(r.center(ancho) for r in roles))
print("-" * (12 + ancho * len(roles)))
for tarea, valores in raci.items():
    print(tarea.ljust(12) + "".join(v.center(ancho) for v in valores))

# Validar reglas
print("\nValidación:")
for tarea, valores in raci.items():
    n_a = sum("A" in v for v in valores)
    n_r = sum("R" in v for v in valores)
    n_c = sum("C" in v for v in valores)
    problemas = []
    if n_a != 1:
        problemas.append(f"tiene {n_a} A (debe haber exactamente 1)")
    if n_r < 1:
        problemas.append("no tiene R (nadie ejecuta)")
    if n_c > len(roles) // 2:
        problemas.append(f"tiene {n_c} C (demasiados consultados)")
    estado = "OK" if not problemas else "; ".join(problemas)
    print(f"  {tarea:<12} {estado}")

# Carga de A/R por rol (detecta cuellos de botella)
print("\nTareas en las que cada rol responde (A):")
for i, rol in enumerate(roles):
    tareas = [t for t, v in raci.items() if "A" in v[i]]
    print(f"  {rol:<16} {len(tareas)} -> {', '.join(tareas) or '—'}")
````

````python
# Simulador de escalamiento según los límites del plan de gobernanza

LIMITES = {
    # variable: (límite del equipo/PM, límite de la PMO); por encima -> patrocinador
    "costo_%":         (5, 10),
    "cronograma_sem":  (1, 3),
}


def escalar(variable, valor):
    equipo, pmo = LIMITES[variable]
    if valor <= equipo:
        return "Decide el equipo / PM"
    if valor <= pmo:
        return "Escalar a la PMO"
    return "Escalar al PATROCINADOR"


situaciones = [
    ("Sobrecosto de licencias",            "costo_%",        3),
    ("Retraso por proveedor externo",      "cronograma_sem", 2),
    ("Cambio de tecnología de la base",    "costo_%",        14),
    ("Enfermedad de un desarrollador",     "cronograma_sem", 1),
    ("Integración con sistema legado",     "cronograma_sem", 5),
]

print(f"{'Situación':<36}{'Variable':<16}{'Valor':>6}   Decisión")
print("-" * 90)
for nombre, variable, valor in situaciones:
    print(f"{nombre:<36}{variable:<16}{valor:>6}   {escalar(variable, valor)}")
````
