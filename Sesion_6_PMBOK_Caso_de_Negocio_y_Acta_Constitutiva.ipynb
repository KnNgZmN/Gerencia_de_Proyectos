## 1. Estructura de la Guía PMBOK® 8ª Edición

### 1.1 Qué cambia respecto al pasado

Las ediciones previas acumularon **capas de procesos, entradas, herramientas y salidas** que terminaron
siendo difíciles de aplicar en la práctica: mucha estructura, poca guía sobre *cómo decidir*. La 8ª edición
propone un **enfoque simplificado**, orientado a principios y a la entrega de valor, y no a memorizar
matrices de procesos.

```
   EL PASADO                              EL PRESENTE
   ---------                              -----------
   Marañas de procesos              ->    Enfoque simplificado
   Estructura rígida                ->    Principios y adaptación (tailoring)
   Foco en la documentación         ->    Foco en el valor entregado
   "Seguir el proceso"              ->    "Sustentar la decisión"
```

### 1.2 Lo que hay que retener

- La guía es un **marco de referencia adaptable**, no un manual de obligatorio cumplimiento paso a paso.
- El gerente de proyectos **selecciona y ajusta** las prácticas según el contexto, el tipo de proyecto y
  la madurez de la organización.
- La documentación existe para **soportar decisiones**, no para justificar la existencia del proyecto.

## 2. Portafolios, Programas, Proyectos y Operaciones

### 2.1 La gestión de cartera vincula la estrategia con la ejecución

Este es el enunciado clave de la sección. La estrategia organizacional no se ejecuta sola: baja hasta el
trabajo diario a través de una jerarquía.

```
        +---------------------------------------------+
        |            ESTRATEGIA / MISIÓN              |   ¿Hacia dónde vamos?
        +---------------------------------------------+
        |            OBJETIVOS ESTRATÉGICOS           |   ¿Qué debemos lograr?
        +---------------------------------------------+
        |            PORTAFOLIO (cartera)             |   ¿En qué invertimos?
        +---------------------------------------------+
        |            PROGRAMAS                        |   ¿Qué agrupamos por beneficio común?
        +--------------------------+------------------+
        |   PROYECTOS              |   OPERACIONES    |   ¿Qué ejecutamos?
        |   (temporales, únicos)   |   (continuas)    |
        +--------------------------+------------------+
                     ^                      ^
                     +---- RECURSOS ORGANIZACIONALES (finitos) ----+
```

### 2.2 Diferencias que suelen preguntarse en evaluación

| Concepto | Naturaleza | Objetivo | Termina cuando |
|---|---|---|---|
| **Proyecto** | Temporal y único | Producir un resultado o entregable específico | Se cumplen los objetivos o se cancela |
| **Programa** | Conjunto de proyectos relacionados | Obtener beneficios que no se logran gestionándolos por separado | Se materializan los beneficios |
| **Portafolio** | Colección de programas, proyectos y operaciones | Cumplir objetivos estratégicos priorizando la inversión | No termina: se revisa y rebalancea |
| **Operación** | Continua y repetitiva | Sostener el negocio del día a día | No termina mientras exista el negocio |

### 2.3 Por qué importa la restricción de recursos

El portafolio existe porque los **recursos organizacionales son finitos**. Autorizar un proyecto significa
*no* autorizar otro. De ahí que el Caso de Negocio (sección 5) sea una **competencia por inversión**, no
un trámite: se compite contra las demás iniciativas de la cartera.

## 3. Sistema de Entrega de Valor

> **"El valor por unidad de inversión es el indicador definitivo del éxito de un proyecto."**

### 3.1 Cómo leer esta frase

El éxito **no** es únicamente terminar a tiempo, dentro del presupuesto y con el alcance completo
(la clásica triple restricción). Un proyecto puede cumplir las tres cosas y aun así **fracasar**, si el
resultado no genera valor suficiente frente a lo que costó.

```
                         VALOR ENTREGADO
   Éxito  ~  ------------------------------------
                      INVERSIÓN REALIZADA
```

### 3.2 El valor tiene varias formas

| Tipo de valor | Ejemplo |
|---|---|
| Financiero | Ingresos nuevos, ahorro de costos, ROI |
| Operativo | Productividad, reducción de tiempos de ciclo |
| De cliente | Satisfacción, retención, experiencia de usuario |
| Estratégico | Posicionamiento, entrada a un mercado, capacidad instalada |
| De riesgo | Cumplimiento normativo, mitigación de exposición |

### 3.3 Flujo del sistema de valor

```
  Necesidad     ->  Iniciativa   ->  Entregable   ->  Resultado   ->  Beneficio  ->  VALOR
  del negocio       autorizada       (output)         (outcome)       medible       sostenido

  "perdemos        "proyecto de     "módulo de      "los pedidos    "-30 % de     "margen
   pedidos"         automatiz."      captura"         se registran    errores"      recuperado"
                                                      sin error"
```

El gerente de proyectos responde por el **entregable**, pero debe entender y defender toda la cadena:
si el beneficio no aparece, el entregable no sirvió de nada.

## 4. Tutorial de Viabilidad — Análisis paso a paso

El estudio de viabilidad es el análisis que **antecede** al Caso de Negocio y lo alimenta. Se ordena en
tres momentos:

```
   +---------------+      +---------------+      +---------------+
   |   NECESIDAD   |  ->  |   OBJETIVOS   |  ->  |     ÉXITO     |
   +---------------+      +---------------+      +---------------+
   | Identificar   |      | Definir       |      | Establecer    |
   | necesidades y |      | objetivos     |      | criterios de  |
   | oportunidades |      | medibles del  |      | éxito y       |
   | empresariales |      | proyecto      |      | validación    |
   +---------------+      +---------------+      +---------------+
```

### Paso 1 — Necesidad: identificar necesidades y oportunidades empresariales

- ¿Cuál es el **problema real** o la oportunidad no aprovechada?
- ¿Qué evidencia lo sustenta (datos, quejas, pérdidas, indicadores)?
- ¿Qué pasa si **no** se hace nada? (escenario base o *do nothing*)
- ¿A qué objetivo estratégico se conecta?

### Paso 2 — Objetivos: definir objetivos medibles del proyecto

- Traducir la necesidad a objetivos **SMART** (específicos, medibles, alcanzables, relevantes, temporales).
- Cada objetivo debe tener: **métrica, línea base, meta y plazo**.
- Ejemplo: *"Reducir el tiempo de registro de pedidos de 12 min a 4 min en un plazo de 6 meses."*

### Paso 3 — Éxito: establecer criterios de éxito y validación

- ¿Cómo se **verificará** que el objetivo se cumplió y **quién** lo valida?
- Criterios de aceptación de los entregables.
- Umbral mínimo aceptable frente al valor esperado.

### Dimensiones de viabilidad a evaluar

| Dimensión | Pregunta que responde |
|---|---|
| Técnica | ¿Existe la tecnología y la capacidad para hacerlo? |
| Económica / financiera | ¿Los beneficios superan los costos? (VPN, TIR, payback, ROI) |
| Operativa | ¿La organización puede absorber y sostener el cambio? |
| Legal / normativa | ¿Cumple regulación, contratos y políticas internas? |
| De cronograma | ¿Es alcanzable en la ventana de tiempo disponible? |

## 5. Documentos formales: Caso de Negocio y Acta Constitutiva

### 5.1 Caso de Negocio

> **Estudio de viabilidad económica para establecer la validez de los beneficios de un proyecto.**

- Responde a la pregunta: **¿vale la pena hacerlo?**
- Se elabora **antes** de autorizar el proyecto.
- Es el insumo con el que el patrocinador y el comité de portafolio **deciden invertir o no**.
- Su dueño típico es el **patrocinador o el área de negocio**, no el gerente de proyecto.

**Contenido habitual:** necesidad del negocio, alineación estratégica, alternativas evaluadas (incluida la
de no hacer nada), análisis costo-beneficio, indicadores financieros, riesgos de alto nivel y recomendación.

### 5.2 Acta Constitutiva (*Project Charter*)

> **Documento que autoriza formalmente un proyecto y otorga autoridad para aplicar recursos.**

- Responde a la pregunta: **¿queda autorizado y con qué límites?**
- Es el documento que **da existencia formal** al proyecto y **nombra al gerente de proyecto**,
  otorgándole autoridad para aplicar recursos de la organización.
- Lo emite y **firma el patrocinador** (alguien con poder de asignar recursos), no el gerente.

**Contenido habitual:** propósito, objetivos medibles y criterios de éxito, requisitos de alto nivel,
descripción y límites del alcance, riesgos principales, hitos, presupuesto asignado, interesados clave,
criterios de salida y designación del gerente de proyecto con su nivel de autoridad.


### 5.3 Comparación directa (tabla vista en clase)

| Caso de Negocio | Acta Constitutiva |
|---|---|
| Necesidades comerciales | Objetivos medibles |
| Alineación estratégica | Límites de alcance |
| Beneficios financieros | Recursos financieros |
| Análisis de viabilidad | Criterios de salida |

**Cómo leer la tabla:** cada fila muestra el mismo asunto en dos niveles distintos. El Caso de Negocio
plantea el **argumento** (necesidad, estrategia, beneficio esperado, viabilidad); el Acta Constitutiva
lo convierte en **compromiso ejecutable** (objetivo medible, frontera de alcance, presupuesto asignado,
condición de cierre).

### 5.4 Secuencia lógica

```
  Necesidad   ->  Estudio de   ->  CASO DE      ->  Decisión de  ->  ACTA          ->  Ejecución
  del negocio     viabilidad       NEGOCIO          inversión        CONSTITUTIVA      del proyecto
                                   (¿vale la        (portafolio)     (autoriza y
                                    pena?)                            asigna al PM)
```

> **Regla mnemotécnica:** el Caso de Negocio **justifica**; el Acta Constitutiva **autoriza**.

## 6. Apoyos prácticos

Las celdas siguientes son ejercicios de repaso ejecutables. No requieren librerías externas.


````python
# Comparación Caso de Negocio vs Acta Constitutiva

comparacion = [
    ("Necesidades comerciales", "Objetivos medibles"),
    ("Alineación estratégica",  "Límites de alcance"),
    ("Beneficios financieros",  "Recursos financieros"),
    ("Análisis de viabilidad",  "Criterios de salida"),
]

ancho = 28
print(f"{'CASO DE NEGOCIO':<{ancho}} | ACTA CONSTITUTIVA")
print("-" * ancho + "-+-" + "-" * ancho)
for caso, acta in comparacion:
    print(f"{caso:<{ancho}} | {acta}")

print("\nEl Caso de Negocio justifica la inversión; el Acta Constitutiva autoriza el proyecto.")
````

````python
# Checklist de verificación de un Acta Constitutiva

elementos_acta = {
    "Propósito y justificación del proyecto": False,
    "Objetivos medibles y criterios de éxito": False,
    "Requisitos de alto nivel": False,
    "Descripción y límites del alcance": False,
    "Riesgos principales identificados": False,
    "Cronograma de hitos resumido": False,
    "Recursos financieros preaprobados": False,
    "Interesados clave": False,
    "Criterios de salida o cierre": False,
    "Gerente de proyecto designado y su autoridad": False,
    "Patrocinador que firma y autoriza": False,
}

# Cambia a True los elementos que sí tenga el acta que estés revisando
total = len(elementos_acta)
cumplidos = sum(elementos_acta.values())

for elemento, ok in elementos_acta.items():
    print(f"[{'X' if ok else ' '}] {elemento}")

print(f"\nCompletitud: {cumplidos}/{total} ({cumplidos / total:.0%})")
````

````python
# Viabilidad económica: indicadores básicos del Caso de Negocio

def vpn(tasa, flujos):
    """Valor Presente Neto. flujos[0] es la inversión inicial (negativa)."""
    return sum(f / (1 + tasa) ** t for t, f in enumerate(flujos))

def payback(flujos):
    """Periodo de recuperación simple, en periodos."""
    acumulado = 0
    for t, f in enumerate(flujos):
        anterior = acumulado
        acumulado += f
        if t > 0 and acumulado >= 0:
            return t - 1 + (-anterior / f)
    return None

def roi(flujos):
    """Retorno sobre la inversión."""
    inversion = -flujos[0]
    beneficio = sum(flujos[1:]) - inversion
    return beneficio / inversion

# Ejemplo: proyecto de automatización evaluado a 5 años
tasa = 0.12
flujos = [-100_000, 30_000, 35_000, 40_000, 45_000, 50_000]

print(f"Inversión inicial : ${-flujos[0]:,.0f}")
print(f"Tasa de descuento : {tasa:.0%}")
print(f"VPN               : ${vpn(tasa, flujos):,.0f}")
print(f"ROI               : {roi(flujos):.1%}")
print(f"Payback           : {payback(flujos):.1f} años")
print()
print("Decisión: si el VPN es mayor que cero, el proyecto crea valor y es viable económicamente.")
````
