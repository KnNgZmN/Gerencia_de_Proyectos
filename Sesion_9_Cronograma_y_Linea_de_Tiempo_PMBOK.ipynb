# PARTE I — Dominio de Programación

## 1. Dominio de Programación

### 1.1 Reaccionar o diseñar

| Reaccionar al tiempo | Diseñar el tiempo |
|---|---|
| Las fechas se descubren cuando llegan | Las fechas se calculan antes de empezar |
| Las dependencias aparecen como sorpresas | Las dependencias están modeladas en el cronograma |
| Todo parece urgente | Se sabe qué actividades son críticas y cuáles tienen holgura |
| Los retrasos se explican después | Los retrasos se anticipan y se miden contra una línea base |

### 1.2 Los cuatro pasos para elaborar el cronograma

```
   ┌──────────────┐      ┌──────────────┐      ┌──────────────┐      ┌──────────────┐
   │   DEFINIR    │  ->  │  SECUENCIAR  │  ->  │   ESTIMAR    │  ->  │   AJUSTAR    │
   ├──────────────┤      ├──────────────┤      ├──────────────┤      ├──────────────┤
   │ Identificar  │      │ Determinar   │      │ Calcular     │      │ Revisar y    │
   │ y documentar │      │ el orden     │      │ esfuerzo y   │      │ modificar el │
   │ acciones     │      │ lógico del   │      │ duración     │      │ borrador     │
   │ específicas  │      │ trabajo      │      │ requerida    │      │              │
   └──────┬───────┘      └──────────────┘      └──────────────┘      └───────┬──────┘
          ▲                                                                  │
          └────────────────────── ciclo iterativo ───────────────────────────┘
```

Los pasos forman un **ciclo**, no una línea recta: al ajustar el borrador aparecen actividades que faltaban,
dependencias mal planteadas o estimaciones irreales, y se vuelve a definir, secuenciar y estimar.

| Paso | Pregunta | Resultado |
|---|---|---|
| **Definir** | ¿Qué acciones concretas producen cada entregable? | Lista de actividades y de hitos |
| **Secuenciar** | ¿En qué orden lógico deben ejecutarse? | Diagrama de red con relaciones y dependencias |
| **Estimar** | ¿Cuánto esfuerzo y cuánta duración requiere cada una? | Estimaciones de duración |
| **Ajustar** | ¿El borrador es realista y cumple las fechas comprometidas? | Cronograma y línea base aprobados |

## 2. Definir actividades

### 2.1 Definición

> **Actividad: porción de trabajo distinta y programada que se realiza para producir los entregables del
> proyecto.**

### 2.2 De la EDT al cronograma

La sesión anterior terminó en el **paquete de trabajo**, el nivel más bajo de la EDT. Definir actividades es
el paso siguiente: descomponer cada paquete en las acciones necesarias para producirlo.

| | Paquete de trabajo (EDT) | Actividad (cronograma) |
|---|---|---|
| Se nombra con | **Sustantivo** — un entregable | **Verbo** — una acción |
| Ejemplo | *Módulo de autenticación* | *Diseñar el flujo de login*, *Programar el servicio*, *Probar el acceso* |
| Responde | ¿Qué se entrega? | ¿Qué hay que hacer para entregarlo? |
| Se usa para | Estimar costo y asignar responsable | Programar fechas y secuencia |

### 2.3 Atributos de una actividad

| Atributo | Ejemplo |
|---|---|
| Identificador | `D` |
| Nombre (con verbo) | Codificar el backend |
| Paquete de la EDT al que pertenece | 1.3 Módulo de servicios |
| Predecesoras y sucesoras | Predecesora: Diseñar el sistema |
| Tipo de relación y ajuste | Final a Inicio, sin ajuste |
| Duración estimada | 12 días |
| Recursos requeridos | 2 desarrolladores backend |
| Restricciones | Debe terminar antes del inicio del piloto |

### 2.4 Hitos

Un **hito** es un punto significativo del proyecto con **duración cero**: no consume tiempo, marca que algo
se logró. Ejemplos: *diseño aprobado*, *código congelado*, *despliegue en producción*.

## 3. Secuencias y lógica

### 3.1 Método de diagramación por precedencia (PDM)

Las actividades se representan como nodos y las dependencias como flechas. Existen **cuatro relaciones**
posibles entre una actividad predecesora (A) y una sucesora (B):

```
   FS · Final a Inicio                       SS · Inicio a Inicio
   A  ████████┐                              A  ┌███████
              └►██████  B                    B  └►█████████
   B no puede COMENZAR hasta que A           B no puede COMENZAR hasta que A
   haya FINALIZADO                           haya COMENZADO


   FF · Final a Final                        SF · Inicio a Final
   A  ████████┐                              A      ┌███████
   B    ██████┘                              B  ████┘
   B no puede FINALIZAR hasta que A          B no puede FINALIZAR hasta que A
   haya FINALIZADO                           haya COMENZADO
```

| Relación | Regla | Ejemplo en software | Frecuencia |
|---|---|---|---|
| **FS** Final a Inicio | La sucesora no puede comenzar hasta que la predecesora haya finalizado | No se prueba un módulo hasta terminar de codificarlo | La más usada |
| **SS** Inicio a Inicio | La sucesora no puede comenzar hasta que la predecesora haya comenzado | La documentación arranca cuando empieza la codificación | Frecuente |
| **FF** Final a Final | La sucesora no puede finalizar hasta que la predecesora haya finalizado | La documentación no se cierra hasta terminar el despliegue | Ocasional |
| **SF** Inicio a Final | La sucesora no puede finalizar hasta que la predecesora haya comenzado | El sistema antiguo no se apaga hasta que arranque el nuevo | Muy rara |

### 3.2 Tipos de dependencias

La relación dice **cómo** se conectan dos actividades; el tipo de dependencia dice **por qué** existe esa
conexión y **quién la controla**.

| Tipo | Definición | Ejemplo | ¿Se puede cambiar? |
|---|---|---|---|
| **Obligatoria** | Inherente a la naturaleza del trabajo | No se despliega código que no existe | No: es lógica dura |
| **Discrecional** | Basada en la lógica preferida o en buenas prácticas | Terminar el diseño antes de codificar | Sí: es la primera candidata para acortar el cronograma |
| **Interna** | Dentro del control del proyecto | El equipo de pruebas espera al de desarrollo | Sí, con decisiones del propio equipo |
| **Externa** | Fuera del control del equipo | Aprobación del cliente, entrega de un proveedor | No directamente: se gestiona como riesgo |

Los tipos se **combinan en dos ejes**:

|  | **Interna** | **Externa** |
|---|---|---|
| **Obligatoria** | Probar después de codificar | Desplegar solo cuando el proveedor habilite el servidor |
| **Discrecional** | Documentar en paralelo con la codificación | Esperar la revisión del cliente antes de integrar |

> **Por qué importa:** cuando hay que acortar un cronograma, las dependencias **discrecionales internas** son
> las primeras que se revisan, porque dependen solo de decisiones del equipo.

## 4. Adelantos y retrasos

### 4.1 Definiciones

| Concepto | Definición | Efecto en la relación | Notación habitual |
|---|---|---|---|
| **Adelanto** (*lead*) | Tiempo en que una sucesora puede **adelantarse** respecto a su predecesora | Acerca o superpone las actividades | Ajuste negativo: `FS − 3 días` |
| **Retraso** (*lag*) | Tiempo de **espera** obligatorio entre la predecesora y la sucesora | Separa las actividades | Ajuste positivo: `FS + 2 días` |

```
   SIN AJUSTE (FS)       A  ████████
                                   ██████  B

   CON ADELANTO (FS −3)  A  ████████
                                ██████  B        B empieza 3 días antes de que A termine

   CON RETRASO (FS +2)   A  ████████
                                     ██████  B   B espera 2 días después de que A termina
```

### 4.2 Ejemplos

- **Adelanto de dos semanas en una relación Final a Inicio:** en una obra, el paisajismo puede empezar dos
  semanas **antes** de terminar la lista de pendientes, porque ambos trabajos no se estorban.
- **Retraso en software:** tras las pruebas de integración hay que **esperar** a que el cliente agende la
  revisión. Nadie trabaja en ese intervalo, pero el tiempo transcurre y debe estar en el cronograma.

> **Error común:** confundir un retraso con una actividad. El retraso **no consume recursos**; una actividad
> sí. Si durante la espera alguien trabaja, eso es una actividad y se modela como tal.

### 4.3 El ADN del cronograma

La tabla vista en clase resume toda la lógica de un cronograma en cinco columnas:

| ID | Actividad | Predecesora | Relación | Ajuste |
|---|---|---|---|---|
| 1 | Diseño | Ninguna | N/A | 0 |
| 2 | Código | 1 | FS | 0 |
| 3 | Pruebas | 2 | FS | Retraso |

Con solo estas columnas más la **duración**, el cronograma completo se puede calcular. Eso es exactamente lo
que hace el tutorial de la siguiente sección.

## 5. Tutorial de software

### 5.1 Los cuatro pasos del ciclo de desarrollo

```
   ●─────────────────────●─────────────────────●─────────────────────●
   PASO 1                PASO 2                PASO 3                PASO 4
   Diseño del sistema    Codificación          Pruebas unitarias     Despliegue
```

El ejemplo de clase se amplía a **10 actividades** para que aparezcan las cuatro relaciones, adelantos,
retrasos y los cuatro tipos de dependencias. Las celdas siguientes calculan el cronograma completo con el
**método de la ruta crítica**:

- **Pasada hacia adelante:** inicio y fin más tempranos de cada actividad.
- **Pasada hacia atrás:** inicio y fin más tardíos sin retrasar el proyecto.
- **Holgura:** cuánto puede atrasarse una actividad sin mover la fecha final.
- **Ruta crítica:** las actividades con holgura cero. Si una de ellas se atrasa, se atrasa todo el proyecto.


````python
# El ADN del cronograma: actividades y dependencias

# id: (nombre, duración en días hábiles)
ACTIVIDADES = {
    "A": ("Levantar requisitos",            5),
    "B": ("Diseñar el sistema",             8),
    "C": ("Configurar ambientes",           4),
    "D": ("Codificar el backend",          12),
    "E": ("Codificar el frontend",         10),
    "F": ("Ejecutar pruebas unitarias",     6),
    "G": ("Ejecutar pruebas de integración", 5),
    "H": ("Obtener aprobación del cliente", 3),
    "I": ("Desplegar en producción",        2),
    "J": ("Redactar documentación técnica", 6),
}

# (predecesora, sucesora, relación, ajuste en días, tipo de dependencia)
# ajuste negativo = adelanto · ajuste positivo = retraso
DEPENDENCIAS = [
    ("A", "B", "FS",  0, "Obligatoria"),
    ("B", "C", "SS",  3, "Discrecional"),
    ("B", "D", "FS",  0, "Obligatoria"),
    ("B", "E", "FS", -3, "Discrecional"),
    ("D", "F", "FS",  0, "Obligatoria"),
    ("E", "F", "FS",  0, "Obligatoria"),
    ("C", "G", "FS",  0, "Obligatoria"),
    ("F", "G", "FS",  0, "Obligatoria"),
    ("G", "H", "FS",  2, "Externa"),
    ("H", "I", "FS",  0, "Obligatoria"),
    ("D", "J", "SS",  5, "Discrecional"),
    ("I", "J", "FF",  0, "Interna"),
]


def describir_ajuste(ajuste):
    if ajuste < 0:
        return f"Adelanto {-ajuste} d"
    if ajuste > 0:
        return f"Retraso {ajuste} d"
    return "0"


print(f"{'ID':<4}{'Actividad':<34}{'Dur':>4}  {'Predecesora':<12}{'Rel':<5}{'Ajuste':<14}Tipo")
print("-" * 92)
for act_id, (nombre, dur) in ACTIVIDADES.items():
    deps = [d for d in DEPENDENCIAS if d[1] == act_id]
    if not deps:
        print(f"{act_id:<4}{nombre:<34}{dur:>4}  {'Ninguna':<12}{'N/A':<5}{'0':<14}—")
    for i, (pre, _, rel, ajuste, tipo) in enumerate(deps):
        izquierda = f"{act_id:<4}{nombre:<34}{dur:>4}  " if i == 0 else " " * 44
        print(f"{izquierda}{pre:<12}{rel:<5}{describir_ajuste(ajuste):<14}{tipo}")

print("\nResumen de la lógica:")
for rel in ("FS", "SS", "FF", "SF"):
    print(f"  Relaciones {rel}: {sum(1 for d in DEPENDENCIAS if d[2] == rel)}")
PLURALES = {"Obligatoria": "obligatorias", "Discrecional": "discrecionales",
            "Interna": "internas", "Externa": "externas"}
for tipo, plural in PLURALES.items():
    print(f"  Dependencias {plural}: {sum(1 for d in DEPENDENCIAS if d[4] == tipo)}")
print(f"  Adelantos: {sum(1 for d in DEPENDENCIAS if d[3] < 0)}  |  Retrasos: {sum(1 for d in DEPENDENCIAS if d[3] > 0)}")
````

````python
# Método de la ruta crítica con las cuatro relaciones de precedencia

def programar(actividades, dependencias):
    dur = {k: v[1] for k, v in actividades.items()}

    # Pasada hacia adelante: inicio más temprano (IT)
    it = {k: 0 for k in actividades}
    for _ in range(len(actividades)):
        for pre, suc, rel, ajuste, _ in dependencias:
            ft_pre = it[pre] + dur[pre]
            minimo = {
                "FS": ft_pre + ajuste,
                "SS": it[pre] + ajuste,
                "FF": ft_pre + ajuste - dur[suc],
                "SF": it[pre] + ajuste - dur[suc],
            }[rel]
            it[suc] = max(it[suc], minimo)
    ft = {k: it[k] + dur[k] for k in actividades}
    fin = max(ft.values())

    # Pasada hacia atrás: fin más tardío (FTa)
    fta = {k: fin for k in actividades}
    for _ in range(len(actividades)):
        for pre, suc, rel, ajuste, _ in dependencias:
            ita_suc = fta[suc] - dur[suc]
            maximo = {
                "FS": ita_suc - ajuste,
                "SS": ita_suc - ajuste + dur[pre],
                "FF": fta[suc] - ajuste,
                "SF": fta[suc] - ajuste + dur[pre],
            }[rel]
            fta[pre] = min(fta[pre], maximo)
    ita = {k: fta[k] - dur[k] for k in actividades}
    holgura = {k: ita[k] - it[k] for k in actividades}
    return it, ft, ita, fta, holgura, fin


it, ft, ita, fta, holgura, fin = programar(ACTIVIDADES, DEPENDENCIAS)
criticas = [k for k in ACTIVIDADES if holgura[k] == 0]

print(f"{'ID':<4}{'Actividad':<34}{'Dur':>4}{'IT':>5}{'FT':>5}{'ITa':>5}{'FTa':>5}{'Holgura':>9}  Crítica")
print("-" * 82)
for k, (nombre, dur) in ACTIVIDADES.items():
    marca = "SÍ" if holgura[k] == 0 else ""
    print(f"{k:<4}{nombre:<34}{dur:>4}{it[k]:>5}{ft[k]:>5}{ita[k]:>5}{fta[k]:>5}{holgura[k]:>9}  {marca}")
print("-" * 82)
print(f"Duración total del proyecto: {fin} días hábiles")
print(f"Ruta crítica: {' → '.join(criticas)}")
print("\nIT/FT = inicio y fin más tempranos · ITa/FTa = inicio y fin más tardíos")
````

````python
# Diagrama de Gantt en texto
#   █ actividad crítica   ▓ actividad con holgura   ░ holgura disponible

ESCALA = 1  # caracteres por día

print(" " * 37 + "".join(f"{d:<5}" for d in range(0, fin + 1, 5)))
print(" " * 37 + "".join("|    " for _ in range(0, fin + 1, 5)))
for k, (nombre, dur) in ACTIVIDADES.items():
    simbolo = "█" if holgura[k] == 0 else "▓"
    barra = " " * it[k] + simbolo * dur + "░" * holgura[k]
    print(f"{k}  {nombre:<34}{barra}")

print(f"\nFin del proyecto: día {fin}")
con_holgura = [f"{k} ({holgura[k]} d)" for k in ACTIVIDADES if holgura[k] > 0]
print(f"Actividades con holgura: {', '.join(con_holgura)}. Pueden atrasarse sin mover la fecha final.")
print("J es crítica aunque no es la más larga: su relación FF la obliga a terminar con el despliegue.")
````

### 5.2 Escenarios: ¿qué pasa si cambian los adelantos y retrasos?

Una de las lecciones más importantes del dominio de programación: **no todo ajuste mueve la fecha final**.
Solo los cambios sobre la **ruta crítica** alteran la duración del proyecto.


````python
# Análisis de escenarios sobre adelantos y retrasos

def escenario(descripcion, cambios):
    """cambios: {(predecesora, sucesora): nuevo_ajuste}"""
    deps = [(p, s, r, cambios.get((p, s), a), t) for p, s, r, a, t in DEPENDENCIAS]
    *_, fin_nuevo = programar(ACTIVIDADES, deps)
    delta = fin_nuevo - fin
    efecto = "sin cambio" if delta == 0 else (f"{delta:+d} días")
    print(f"  {descripcion:<66}{fin_nuevo:>4} días  ({efecto})")


print(f"Línea base: {fin} días\n")
escenario("1. Quitar el adelanto del frontend (B→E: FS −3 → FS 0)",        {("B", "E"): 0})
escenario("2. El cliente tarda más en agendar la revisión (G→H: +2 → +5)",  {("G", "H"): 5})
escenario("3. Adelantar las pruebas unitarias 2 días (D→F: FS 0 → FS −2)",   {("D", "F"): -2})
escenario("4. Retrasar el inicio de ambientes (B→C: SS +3 → SS +15)",        {("B", "C"): 15})

print("""
Lectura de los escenarios:
  1 y 4 no mueven la fecha final porque afectan actividades con holgura (E y C).
  2 atrasa todo el proyecto: la aprobación del cliente es una dependencia EXTERNA
    y está en la ruta crítica, por eso debe gestionarse como riesgo.
  3 acorta el proyecto: un adelanto solo sirve si se aplica sobre la ruta crítica.""")
````

# PARTE II — Línea del tiempo: cómo se usa el PMBOK en un proyecto

## 7. El marco de referencia

La Guía PMBOK® 8 organiza la gestión en dos ejes que se cruzan a lo largo del tiempo:

| Eje | Qué es | Componentes |
|---|---|---|
| **Áreas de enfoque** | *Cuándo* ocurre la gestión: el recorrido del proyecto | Iniciar · Planificar · Ejecutar · Monitorear y controlar · Cerrar |
| **Dominios de desempeño** | *Qué* se gestiona durante todo ese recorrido | Gobernanza · Alcance · Cronograma · Finanzas · Interesados · Recursos · Riesgo |

Las áreas de enfoque son **secuenciales con solapamiento**; los dominios de desempeño son **transversales**:
no se "hacen" una vez, están activos durante todo el proyecto con distinta intensidad.

### Caso aplicado

Se usa como hilo conductor el proyecto **AsistIA** (asistente de IA para servicio al cliente) desarrollado
en el trabajo de implementación en Jira, adaptado a un ciclo de vida **híbrido**: planeación predictiva al
inicio y ejecución ágil en seis sprints.

## 9. Diagrama de fases

Cada fila es una fase y cada columna una semana. Las barras se **solapan** a propósito: el monitoreo y control acompaña toda la ejecución, y la planificación del cronograma y del presupuesto ocurren en paralelo.

```
                                          1    6    11   16   21   26   
                                          |    |    |    |    |    |    
F0  Alineación con estrategia y portafolio··
F1  Caso de negocio y viabilidad           ▒▒▒
F2  Acta constitutiva e interesados          ▒▒
```

`●` dominio principal en esa área de enfoque · `○` dominio de apoyo

| **Alcance** | ○ | ● | ● | ● | ● |
| **Cronograma** | ○ | ● | ● | ● | ○ |
| **Finanzas** | ● | ● | ○ | ● | ● |
| **Interesados** | ● | ● | ● | ● | ● |
| **Recursos** | ○ | ● | ● | ● | ● |
| **Riesgo** | ● | ● | ● | ● | ○ |

Dos lecturas de la matriz:

- **Interesados** es el único dominio principal en las cinco áreas: el proyecto nunca deja de gestionar
  expectativas.
- **Planificar** es la única columna llena de `●`: todos los dominios se planifican, aunque después cada uno
  tenga su propio ritmo.


````python
# Línea del tiempo del PMBOK como datos: validación y resumen

FASES = [
    # (id, nombre, semana inicio, semana fin, área de enfoque, dominios, sesión del curso)
    ("F0",  "Alineación con estrategia y portafolio", 1,  2,  "Pre-proyecto",           ["Gobernanza", "Finanzas"],                         "S6"),
    ("F1",  "Caso de negocio y viabilidad",           2,  4,  "Iniciar",                ["Finanzas", "Interesados"],                        "S6"),
    ("F2",  "Acta constitutiva e interesados",        4,  5,  "Iniciar",                ["Gobernanza", "Interesados"],                      "S6"),
    ("F3",  "Plan de gobernanza y RACI",              5,  6,  "Planificar",             ["Gobernanza", "Recursos"],                         "S7"),
    ("F4",  "Alcance: EDT, diccionario y EDV",        6,  7,  "Planificar",             ["Alcance"],                                        "S8"),
    ("F5",  "Cronograma y ruta crítica",              7,  8,  "Planificar",             ["Cronograma"],                                     "S9"),
    ("F6",  "Presupuesto, riesgos y recursos",        7,  8,  "Planificar",             ["Finanzas", "Riesgo", "Recursos"],                 "Proyecto"),
    ("F7",  "Ejecución en 6 sprints",                 9,  20, "Ejecutar",               ["Alcance", "Cronograma", "Recursos", "Interesados"], "Proyecto"),
    ("F8",  "Monitoreo y control continuo",           9,  21, "Monitorear y controlar", ["Gobernanza", "Alcance", "Cronograma", "Finanzas", "Interesados", "Recursos", "Riesgo"], "S7"),
    ("F9",  "Piloto y aceptación",                    20, 21, "Monitorear y controlar", ["Alcance", "Interesados"],                         "S6"),
    ("F10", "Cierre y lecciones aprendidas",          21, 22, "Cerrar",                 ["Gobernanza", "Recursos"],                         "Proyecto"),
    ("F11", "Medición de beneficios",                 22, 30, "Post-proyecto",          ["Finanzas", "Interesados"],                        "S6 · S8"),
]

DOMINIOS = ["Gobernanza", "Alcance", "Cronograma", "Finanzas", "Interesados", "Recursos", "Riesgo"]
AREAS = ["Iniciar", "Planificar", "Ejecutar", "Monitorear y controlar", "Cerrar"]

# 1. Semanas de actividad de cada dominio
print("Semanas en que cada dominio de desempeño está activo:")
for dominio in DOMINIOS:
    semanas = set()
    for _, _, ini, fin_s, _, doms, _ in FASES:
        if dominio in doms:
            semanas.update(range(ini, fin_s + 1))
    linea = "".join("■" if s in semanas else "·" for s in range(1, 31))
    print(f"  {dominio:<12}{linea}  {len(semanas):>2} semanas")
print(f"  {'':<12}" + "".join(str(s % 10) for s in range(1, 31)) + "  (semana)")

# 2. Cobertura: toda área de enfoque tiene al menos una fase
print("\nFases por área de enfoque:")
for area in AREAS:
    fases = [nombre for _, nombre, _, _, a, _, _ in FASES if a == area]
    estado = "OK" if fases else "SIN FASES"
    print(f"  {area:<24}{len(fases)} fase(s)  [{estado}]")

# 3. Continuidad: no hay semanas vacías entre la primera y la última fase
ocupadas = set()
for _, _, ini, fin_s, _, _, _ in FASES:
    ocupadas.update(range(ini, fin_s + 1))
huecos = [s for s in range(1, 31) if s not in ocupadas]
print(f"\nSemanas sin ninguna fase activa: {huecos if huecos else 'ninguna'}")

# 4. Recorrido por las sesiones del curso
print("\nRecorrido del curso a lo largo del proyecto:")
for fid, nombre, ini, fin_s, area, _, sesion in FASES:
    print(f"  Sem {ini:>2}–{fin_s:<2} {sesion:<9}{nombre}")
````

## Síntesis

### Parte I — Dominio de programación

1. El cronograma se construye en un **ciclo de cuatro pasos**: definir, secuenciar, estimar y ajustar.
2. Una **actividad** es trabajo programado con verbo; nace de descomponer los paquetes de trabajo de la EDT.
3. Las **cuatro relaciones de precedencia** (FS, SS, FF, SF) dicen cómo se conectan las actividades; los
   **cuatro tipos de dependencia** (obligatoria, discrecional, interna, externa) dicen por qué y quién las controla.
4. Los **adelantos** superponen actividades y los **retrasos** imponen esperas que no consumen recursos.
5. La **ruta crítica** determina la duración del proyecto: ajustar actividades con holgura no mueve la fecha final.

### Parte II — El PMBOK a lo largo del proyecto

6. El PMBOK no se aplica en orden de capítulos: las **áreas de enfoque** marcan el recorrido en el tiempo y
   los **dominios de desempeño** están activos durante todo ese recorrido.
7. Lo visto en el curso encaja en una secuencia: **valor y caso de negocio** (S6) → **gobernanza** (S7) →
   **alcance** (S8) → **cronograma** (S9) → ejecución y control con la herramienta.
8. El proyecto no termina en el cierre: termina cuando se **verifican los beneficios** que justificaron la
   inversión.
